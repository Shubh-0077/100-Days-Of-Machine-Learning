## 🔢 Numerical Transformation (Numerical Columns)
 There are mainly two types of transformations:

### 1️⃣ Discretization (Binning / Bracket Binning)
 - Converts continuous numerical values into discrete intervals (bins).
 - Helps in simplifying data and handling non-linear relationships.
 - Example: Age → [0-18], [19-35], [36-60], [60+]

### 2️⃣ Binarization
 - Converts numerical values into binary form (0 or 1) based on a threshold.
 - Useful for presence/absence or yes/no type features.
 - Example: Salary > 50000 → 1, else → 0

```python
import numpy as np 
import pandas as pd 

from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.metrics import accuracy_score

from sklearn.tree import DecisionTreeClassifier

from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','SibSp','Parch','Survived'])
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df['Family'] = df['SibSp'] + df['Parch']
```

```python
df.sample(3)
```

```python
df.drop(columns=['SibSp','Parch'],inplace= True)
```

```python
df.sample(3)
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

Without Binarization

```python
clf = DecisionTreeClassifier()

clf.fit(x_train,y_train)

y_pred = clf.predict(x_test)

accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(clf,x,y,cv=15,scoring='accuracy'))
```
