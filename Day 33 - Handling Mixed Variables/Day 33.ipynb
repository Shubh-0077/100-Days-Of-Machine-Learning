```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

```python
# lest Check the Unique Value for Number and Cabin Cols
df['number'].unique()
```

```python
df['number'].value_counts().plot(kind='bar') ;
```

```python
#  there are Numbers and Letters in number col . Lets Seprate them
```

```python
# extract numerical part 
df['number_num'] = pd.to_numeric(df["number"],errors='coerce',downcast='integer') 
```

```python
# extract categorical part
df['number_cate'] = np.where(df['number_num'].isnull(),df['number'],np.nan)
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
# Exctracting Cabin Col

df['cabin_num'] = df['Cabin'].str.extract('(\d+)') # captures numerical part
df['cabin_cat'] = df['Cabin'].str[0] # captures the first letter

df.head()
```

```python
df['cabin_cat'].value_counts().plot(kind='bar') ;
```

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])  #Splits the values using delimeter of ( )
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',  #Hanles Errors 
                                   downcast='integer') # Saves Memory 

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])    
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,df['ticket_cat'])
#  using str.isdigit() to check any numeric values in cat col , if there replace it with NaN ( np.nan)


df.head(10)
```
