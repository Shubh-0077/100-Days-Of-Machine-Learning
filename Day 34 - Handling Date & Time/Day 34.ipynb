```python
import numpy as np 
import pandas as pd 
```

```python
date = pd.read_csv('orders.csv')
time = pd.read_csv('messages.csv')
```

## Working w Dates

```python
date = date.drop(columns=['product_id','city_id','orders'])
date.head()
```

```python
# First Always Check the data type of date col , So u can use all the date features
date.info()
```

```python
# converting str(object) to datetime data type 
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

### Extract Year

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

### Extract Month

```python
date['date_month'] = date['date'].dt.month

date.sample(5)
```

### Extract Days

```python
date['date_day'] = date['date'].dt.day

date.sample(5)
```

```python
# Day of the Week
date['date_week'] = date['date'].dt.dayofweek

date.sample(5)
```

```python
# Day of the week name 
date['date_dow_name'] = date['date'].dt.day_name()

date.sample(5)
```

```python
# is weekend?

date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)   # Used Numpy Logic for extraction
date.sample(5)
```

### Extract Week of the Year

```python
date['Year_date_week'] = date['date'].dt.isocalendar().week
date.sample(5)
```

```python
# by mistakly i have extracted 2 duplicate cols , thats why i am dropping them
date = date.drop(columns=['date_week','date_day_name'])
```

### Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter
date.sample(5)
```

### Extract Semester 

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)
```

```python
date.sample(5)
```

### Extract Time elapsed between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
# we get the diff between current time the Value time
today - date['date']
```

```python
#  To get the no. of days passed
(today - date['date']).dt.days
```

```python
# Weeks passed
np.round((today -date['date']) / np.timedelta64(1, 'W'),0)
```

```python
# Convert obj to datetime
time['date'] = pd.to_datetime(time['date'])
time.info()
```

### Time difference

```python
today - time['date']
```

```python
# in seconds

(today - time['date'])/np.timedelta64(1,'s')
```

```python
# in minutes

(today - time['date'])/np.timedelta64(1,'m')
```

```python
# in hours

(today - time['date'])/np.timedelta64(1,'h')
```
