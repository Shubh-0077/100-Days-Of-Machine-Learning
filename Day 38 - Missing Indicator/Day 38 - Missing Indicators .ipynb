```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer ,MissingIndicator  

# MissingIndicator will make a new col where , if Value is NaN -> True , if value is present -> False
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
x_train.head()
```

```python
# Without having Missing Imputer 
si = SimpleImputer()
x_train_trf = si.fit_transform(x_train)
x_test_trf = si.transform(x_test)
```

```python
x_train_trf
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
clf = LogisticRegression()

clf.fit(x_train_trf,y_train)          # While Traing mode only using Training Dataset 
y_pred = clf.predict(x_test_trf)      # Predict on x Test 

accuracy_score(y_test,y_pred)
```

```python
# Lets Apply Missing Imputer and see if our model performces well or not 
mi = MissingIndicator()

mi.fit(x_train)
```

```python
mi.features_

# MissingIndicator is used to identify which columns contain missing values.
# [0] means the first col has missing values (Age col)
```

```python
# lets create missing indicator a new col
x_train_missing = mi.transform(x_train)
x_test_missing = mi.transform(x_test)
```

```python
x_train_missing
```

```python
x_test_missing
```

```python
# Making a new col in the df
x_train['Age_NA'] = x_train_missing
x_test['Age_NA'] = x_test_missing
```

```python
x_train.tail()
```

```python
x_test.tail()
```

```python
# now lets train our model once again and see the results 
si1 = SimpleImputer()

x_train_trf2 = si1.fit_transform(x_train)
x_test_trf2 = si1.transform(x_test)
```

```python
clf1 = LogisticRegression()

clf1.fit(x_train_trf2,y_train)
y_pred2 = clf1.predict(x_test_trf2)

accuracy_score(y_test,y_pred2)
```

```python
# Result : 61% -> 63%  . Our model has incereased slightly 
```

### Short Cut Way

This method was so popular ,That SkLearn included in the parameters   
its become more easier now 

```python
si = SimpleImputer(add_indicator=True) 

# add_indicator=True does the same job what Missing indicator was doing 
```

```python
x_train = si.fit_transform(x_train)
x_test = si.transform(x_test)
```

```python
clf = LogisticRegression()
clf.fit(x_train_trf2,y_train)
y_pred = clf.predict(x_test_trf2)
```

```python
accuracy_score(y_test,y_pred)  
# Got same result as Missing indicator
```
