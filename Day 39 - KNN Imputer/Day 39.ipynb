# KNN Imputer
https://scikit-learn.org/stable/modules/generated/sklearn.impute.KNNImputer.html 
## Advantages
- Performs better than mean, median, and mode imputation in most cases.
- Preserves relationships and patterns between features.
- Uses nearest neighboring data points for more realistic imputations.

## Disadvantages
- Computationally expensive because it performs many distance calculations.
- Slow on large datasets.
- Requires more memory/storage in production since training data is needed for neighbor calculations.

## Important Notes
- Default missing values are represented using `np.nan`.
- `weights='uniform'` gives equal importance to neighbors.
- `weights='distance'` gives more importance to closer neighbors using reciprocal distance.

```python
import numpy as np 
import pandas as pd  

from sklearn.model_selection import train_test_split

from sklearn.impute import SimpleImputer ,KNNImputer 

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Pclass','Fare','Survived'])
df.head()
```

```python
df.isnull().mean()*100
```

```python
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
x_train , x_test ,y_train ,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
# Simple Imputer -> Mean 
si = SimpleImputer()

x_train_trf = si.fit_transform(x_train)
x_test_trf = si.transform(x_test)
```

```python
# Training 
lr = LogisticRegression()

lr.fit(x_train_trf,y_train)
y_pred = lr.predict(x_test_trf)

accuracy_score(y_test,y_pred)
```

```python
# Now KNN imputation 
knn = KNNImputer(n_neighbors=5, weights='uniform')

x_train_trf2 = knn.fit_transform(x_train)
x_test_trf2 = knn.transform(x_test)
```

```python
# Training 
lr = LogisticRegression()

lr.fit(x_train_trf2,y_train)
y_pred2 = lr.predict(x_test_trf2)

accuracy_score(y_test,y_pred2)
```
