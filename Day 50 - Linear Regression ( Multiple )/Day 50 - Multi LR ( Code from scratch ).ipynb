### Making our own Linear Regression Class

````python
class MeraLR : 

    def __init__(self):
        self.coef_ = None
        self.int_ = None 

    def fit(self,x_train,y_train):
        
        x_train = np.insert(x_train,0,1,axis=1)       # Note : A 

        # To calculate coeff 
        betas = np.linalg.inv(np.dot(x_train.T,x_train)).dot(x_train.T).dot(y_train)    # Note : B
        self.int_ = betas[0]
        self.coef_ = betas[1:]

    def predict(self,x_test): 
        y_pred = np.dot(x_test,self.coef_) + self.int_      # Note : C
        return y_pred
        
````

# Notes 
## A ) Why We Add a Column of 1s in Linear Regression

The linear regression equation is:

y = b₀ + b₁X₁ + b₂X₂ + ... + bₙXₙ

By inserting a column of 1s into `X_train`, we convert the intercept term (`b₀`) into a coefficient so it can be calculated using matrix multiplication through the Normal Equation.

```python
X_train = np.insert(X_train, 0, 1, axis=1)
```
https://numpy.org/doc/stable/reference/generated/numpy.insert.html

### Why This Is Necessary

- **Mathematical Integration:**  
  The intercept (`b₀`) is a constant. Multiplying it by 1 allows it to behave like a normal coefficient inside matrix operations.

- **Vectorized Calculation:**  
  This allows the Normal Equation:  

  `(XᵀX)⁻¹XᵀY`    to calculate both intercept and slope coefficients together in one operation.

- **Avoiding Origin Bias:**  
  Without this column, the regression line would be forced to pass through `(0,0)`, which can produce inaccurate predictions.

## B ) 

$Beta(\beta) = (X^T X)^{-1} X^T y$

## C ) The prediction equation is:

ŷ = Xβ + b

Where:

- `np.dot(x_test, self.coef_)` → Weighted sum of input features
- `self.int_` → Intercept / Bias term
- `y_pred` → Final prediction values generated by the model
