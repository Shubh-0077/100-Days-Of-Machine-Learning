### Code from Scracth

````python
class GDreg : 

    def __init__ (self,learning_rate = 0.01,epochs = 100): 
        self.m = None
        self.b = None
        self.lr =learning_rate 
        self.epochs = epochs

    def fit(self,x_train,y_train) : 
        # Init coeff_ 
        self.m = np.ones(x_train.shape[1])              # ----- (A)
        self.b = 0 

        for i in range(self.epochs): 
            # update all m & b 
            y_hat = np.dot(x_train,self.m) + self.b           # ------(B)
            
            b_deri = -2 * np.mean(y_train - y_hat)            # ------(C)
            self.b = self.b - (self.lr * b_deri)              

            m_deri = -2 * np.dot((y_train - y_hat),x_train) / x_train.shape[0]             # -------(D)
            self.m = self.m - (self.lr * m_deri)
            
        print(self.b,self.m)
    
    def predict(self,x_test) : 
        return np.dot(x_test,self.m) + self.b
````

## Notes : 

### (B) Multiple Linear Regression Prediction Formula

To calculate the predicted value $(\hat{y})$ in Multiple Linear Regression, we use:

$$
\hat{y}
=
\beta_0
+
\beta_1x_1
+
\beta_2x_2
+
\beta_3x_3
+\dots+
\beta_nx_n
$$
----
Where:
- $(\beta_0)$ → intercept/bias term
- $(\beta_1, \beta_2, \beta_3)$ → coefficients/weights
- $(x_1, x_2, x_3)$ → feature values

---

#### Matrix Representation

Instead of writing long equations manually, we convert the formula into matrix/vector form:

$$
\hat{y}
=
\beta_0
+
\begin{bmatrix}
x_{11} & x_{12} & x_{13}
\end{bmatrix}
\begin{bmatrix}
\beta_1 \\
\beta_2 \\
\beta_3
\end{bmatrix}
$$
---
This makes computations:
- faster,
- cleaner,
- and scalable for large datasets.

---

#### Converting Mathematics → Python (NumPy)

Mathematical formula:

$$
\hat{y} = X\beta + \beta_0
$$

NumPy implementation:

```python
y_pred = np.dot(X_train, coef_) + intercept_
```

Where:
- `X_train` → feature matrix
- `coef_` → coefficients (\(\beta\))
- `intercept_` → bias/intercept (\(\beta_0\))
- `np.dot()` → performs matrix multiplication
    
  


### (C) Simplified Gradient Formula for Intercept $(\beta_0)$

Original formula:

$$
-\frac{2}{n} \sum_{i=1}^{n}(y_i - \hat{y}_i)
=
\frac{\partial L}{\partial \beta_0}
$$

Since:

$$
\frac{\sum (y_i - \hat{y}_i)}{n}
=
\text{mean}(y_i - \hat{y}_i)
$$

we can simplify the formula as:

$$
-2 \cdot \text{mean}(y_i - \hat{y}_i)
=
\frac{\partial L}{\partial \beta_0}
$$

---

### NumPy Implementation

```python
-2 * np.mean(y - y_pred)
```

This computes the gradient (slope) of the loss function with respect to the intercept term $(\beta_0)$.
