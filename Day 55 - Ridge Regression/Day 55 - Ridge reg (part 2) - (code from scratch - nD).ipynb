## Ridge Regression Mathematical Formulation (N-Dimensional Case)

---
For better understanding watch this befor: https://youtu.be/NU37mF5q8VE?si=UANkQlS_sK-cWxGB&t=212 

### Step 1: Linear Regression Equation

For multiple features:

$$
\hat{Y} = X\beta
$$

Where:

- $(X)$ = Feature Matrix
- $(\beta)$ = Coefficient Vector
- $(\hat{Y})$ = Predicted Output

---

### Step 2: Ordinary Least Squares (OLS) Loss Function

Linear Regression tries to minimize:

$$
L(\beta)
=
(Y - X\beta)^T(Y - X\beta)
$$

This is the Residual Sum of Squares (RSS).

Goal:

$$
\min_{\beta}
(Y - X\beta)^T(Y - X\beta)
$$

---

### Problem with OLS

When:
- features are highly correlated,
- number of features becomes large,
- or overfitting occurs,

OLS coefficients can become very large.

Large coefficients increase model complexity and variance.

---

### Step 3: Ridge Regression Loss Function

Ridge adds a penalty term:

$$
L(\beta)
=
(Y - X\beta)^T(Y - X\beta)
+
\lambda \beta^T\beta
$$
-
or

$$
L(\beta)
=
RSS
+
\lambda \sum_{j=1}^{n}\beta_j^2
$$
-
Where:

- $(\lambda)$ = Regularization Parameter
- $(\beta_j)$ = coefficients

---

### Main Intuition

Ridge says:

> Fit the data properly, but also keep coefficients small.

---

### Step 4: Expand the Loss Function

Expanding:

$$
L(\beta)
=
(Y^TY)
-
2\beta^TX^TY
+
\beta^TX^TX\beta
+
\lambda\beta^T\beta
$$
---

### Step 5: Differentiate w.r.t. $(\beta)$

Taking derivative:

$$
\frac{\partial L}{\partial \beta}
=
-2X^TY
+
2X^TX\beta
+
2\lambda\beta
$$
---

### Step 6: Set Derivative = 0

For minimum loss:

$$
-2X^TY
+
2X^TX\beta
+
2\lambda\beta
=
0
$$
-
Divide by 2:

$$
-X^TY
+
X^TX\beta
+
\lambda\beta
=
0
$$
---

### Step 7: Factor Out \(\beta\)

$$
(X^TX + \lambda I)\beta
=
X^TY
$$
-
Where:

$$
I
=
Identity\ Matrix
$$
---

### Step 8: Final Ridge Regression Solution

Multiply inverse on both sides:

$$
\beta
=
(X^TX + \lambda I)^{-1}
X^TY
$$

This is the final Ridge Regression equation.
---

## Comparison

#### Linear Regression

$$
\beta
=
(X^TX)^{-1}
X^TY
$$

#### Ridge Regression

$$
\beta
=
(X^TX + \lambda I)^{-1}
X^TY
$$
-
Notice:

$$
\lambda I
$$

is the additional Ridge penalty term.

---

### Role of Lambda (\(\lambda\))

#### Small \(\lambda\)

$$
\lambda \rightarrow 0
$$

Ridge behaves almost like Linear Regression.

---

#### Large \(\lambda\)

$$
\lambda \rightarrow \infty
$$

- coefficients shrink heavily,
- model becomes simpler,
- variance decreases,
- bias increases.

---

### Geometric Intuition

Linear Regression:

> Finds the point with minimum RSS.

Ridge Regression:

> Finds the point with minimum RSS while restricting coefficient size.

Because of this restriction:

- overfitting decreases,
- coefficients become stable,
- model generalizes better.

---

### Final Formula to Remember

$$
\boxed{
\beta
=
(X^TX + \lambda I)^{-1}
X^TY
}
$$
-
This is the N-dimensional closed-form solution of Ridge Regression.

# Matrix Derivative Reference Table

While differentiating Ridge Regression, the following matrix derivative rules are commonly used:

| Scalar Derivative | Result | Vector/Matrix Derivative | Result |
|----------|----------|----------|----------|
| $(f(x)=bx)$ | $(b)$ | $(f(X)=X^TB)$ | $(B)$ |
| $(f(x)=bx^2)$ | $(2bx)$ | $(f(X)=X^TX)$ | $(2X)$ |
| $(f(x)=x^2)$ | $(2x)$ | $(f(X)=X^TAX)$ | $(2AX)$ |
| $(f(x)=bx^2)$ | $(2bx)$ | $(f(X)=BX)$ | $(B)$ |

> These rules help us differentiate matrix equations without expanding every term manually.

---

# Matrix Notation Explained

Before understanding Ridge Regression, it is important to understand the matrices involved.

### Feature Matrix (\(X\))

$$
X =
\begin{bmatrix}
x_{11} & x_{12} & x_{13} \\
x_{21} & x_{22} & x_{23} \\
x_{31} & x_{32} & x_{33}
\end{bmatrix}
$$

Where:
- Rows = observations (samples)
- Columns = features

Shape:

$$
(m \times n)
$$

where:
- \(m\) = number of rows
- \(n\) = number of features

---

### Target Vector (\(Y\))

$$
Y =
\begin{bmatrix}
y_1 \\
y_2 \\
y_3
\end{bmatrix}
$$

Shape:

$$
(m \times 1)
$$

---

### Coefficient Vector (\(\beta\))

$$
\beta =
\begin{bmatrix}
\beta_1 \\
\beta_2 \\
\beta_3
\end{bmatrix}
$$

Shape:

$$
(n \times 1)
$$

Each coefficient represents the weight of a feature.

---

### Prediction Equation

The complete multiple linear regression equation can be written as:

$$
\hat{Y}=X\beta
$$

Shape Check:

$$
(m \times n)(n \times 1)
=
(m \times 1)
$$

which gives one prediction for each row.

---

### Matrix Multiplication Intuition

For one row:

$$
\hat{y}
=
\beta_1x_1
+
\beta_2x_2
+
\beta_3x_3
$$

For all rows together:

$$
\hat{Y}=X\beta
$$

This is simply the vectorized version of Multiple Linear Regression.

---

### Ridge Regression Formula

After differentiation and simplification, we obtain:

$$
\beta
=
(X^TX+\lambda I)^{-1}X^TY
$$

Where:

- \(X^TX\) → feature correlation matrix
- \(I\) → identity matrix
- \(\lambda\) → regularization parameter
- \(X^TY\) → relationship between features and target

The term:

$$
\lambda I
$$

is what makes Ridge Regression different from Ordinary Least Squares (OLS). 


# Using sklearn

````python
import numpy as np 

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split 
from sklearn.metrics import r2_score 
from sklearn.linear_model import Ridge
````

````python
x,y = load_diabetes(return_X_y=True)
````

````python
x_train ,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=4)
````

````python
rid = Ridge(alpha=0.1,solver='cholesky')
````

solver : 
https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Ridge.html

````python
rid.fit(x_train,y_train)
````

````python
y_pred = rid.predict(x_test)

r2_score(y_test,y_pred)
````

````python
print('m:',rid.coef_)
print('b:',rid.intercept_)
````

# Code from scratch

````python
class MeraRidge: 

    def __init__(self,alpha=0.1): 
        self.alpha = alpha
        self.m = None 
        self.b = None

    def fit(self,x_train,y_train):
         
        x_train = np.insert(x_train,0,1,axis=1)             # ------- (A)
        I = np.identity(x_train.shape[1])                   # ------- (B)
        I[0][0] = 0                                      
        result = np.linalg.inv(np.dot(x_train.T,x_train) + self.alpha * I).dot(x_train.T).dot(y_train)               # ------- (C)

        self.b = result[0]
        self.m = result[1:]
    
    def predict(self,y_test): 
        return np.dot(x_test,self.m) + self.b
````

````python
reg = MeraRidge()
reg.fit(x_train,y_train)
y_pred2 = reg.predict(x_test)

r2_score(y_test,y_pred2)
print('m:',reg.m)
print('b:',reg.b)
````

# Notes
----
## A) Adding the Intercept Column

```python
x_train = np.insert(x_train, 0, 1, axis=1)
```

We add a column of **1s** at the beginning of the feature matrix so that the intercept $(\beta_0))$ can be learned using matrix multiplication.

### Mathematical Form

Without intercept:

$$
\hat{Y}=X\beta
$$

With intercept:

$$
\hat{Y}=\beta_0+\beta_1x_1+\beta_2x_2+\cdots+\beta_nx_n
$$

By adding a column of ones:

$$
X=
\begin{bmatrix}
1 & x_{11} & x_{12} \\
1 & x_{21} & x_{22} \\
1 & x_{31} & x_{32}
\end{bmatrix}
$$

the intercept becomes part of the coefficient vector:

$$
\beta=
\begin{bmatrix}
\beta_0 \\
\beta_1 \\
\beta_2
\end{bmatrix}
$$

This allows us to compute all coefficients together using matrix operations.

---

## B) Creating the Identity Matrix

```python
I = np.identity(x_train.shape[1])
```

This creates the Identity Matrix $(I)$ used in the Ridge Regression formula:

$$
\beta=(X^TX+\lambda I)^{-1}X^TY
$$

Example:

$$
I=
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

The identity matrix ensures that the regularization penalty is applied independently to each coefficient.

---

## Note B.1 — Why `I[0][0] = 0`?

```python
I[0][0] = 0
```

After this modification:

$$
I=
\begin{bmatrix}
0 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

### Why do we do this?

The first coefficient corresponds to the intercept $(\beta_0)$.

In Ridge Regression, we usually **do not regularize the intercept** because:

- The intercept only shifts the regression line up or down.
- It does not control model complexity.
- Penalizing it can unnecessarily bias predictions.

So instead of:

$$
\lambda(\beta_0^2+\beta_1^2+\beta_2^2+\cdots)
$$

we use:

$$
\lambda(\beta_1^2+\beta_2^2+\cdots)
$$

Only feature coefficients are penalized.

### Intuition

- Intercept $(\beta_0)$ → not penalized
- Slopes $(\beta_1,\beta_2,\ldots,\beta_n)$ → penalized

---

## C) Ridge Regression Closed Form Solution

```python
result = np.linalg.inv(
    np.dot(x_train.T, x_train) + self.alpha * I
).dot(x_train.T).dot(y_train)
```

This is the direct implementation of the Ridge Regression formula:

$$
\beta=(X^TX+\lambda I)^{-1}X^TY
$$

### Code → Math Mapping

| Code | Mathematical Meaning |
|--------|--------|
| `x_train.T` | $(X^T)$ |
| `np.dot(x_train.T, x_train)` | $(X^TX)$ |
| `self.alpha * I` | $(\lambda I)$ |
| `np.linalg.inv(...)` | $((X^TX+\lambda I)^{-1})$ |
| `.dot(x_train.T)` | $(X^T)$ |
| `.dot(y_train)` | $(Y)$ |

Therefore:

```python
(XᵀX + λI)⁻¹ XᵀY
```

is exactly:

$$
\beta=(X^TX+\lambda I)^{-1}X^TY
$$

---

## Final Output

```python
self.b = result[0]
self.m = result[1:]
```

The coefficient vector returned by Ridge Regression looks like:

$$
\beta=
\begin{bmatrix}
\beta_0 \\
\beta_1 \\
\beta_2 \\
\vdots
\end{bmatrix}
$$

Where:

- `result[0]` → Intercept $(\beta_0)$
- `result[1:]` → Feature coefficients $(\beta_1,\beta_2,\ldots)$

These values are then used for prediction:

$$
\hat{Y}=X\beta+b
$$
