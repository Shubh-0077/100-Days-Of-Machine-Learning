# Ridge Regression using Gradient Descent

## What is Ridge Regression?

Ridge Regression is an extension of Linear Regression that adds a penalty term to reduce overfitting by shrinking large coefficient values.

### Linear Regression Loss

L = Σ(yi - ŷi)²

### Ridge Regression Loss

L = Σ(yi - ŷi)² + λΣ(wj²)

Where:

- yi → Actual value
- ŷi → Predicted value
- wj → Coefficients (weights)
- λ (lambda) → Regularization parameter

---

## Intuition Behind Ridge Regression

Linear Regression only tries to minimize prediction error.

As a result, some coefficients may become very large, causing:

- Overfitting
- High Variance
- Poor Generalization

Ridge Regression solves this by adding a penalty term:

λΣ(wj²)

This penalty pushes coefficients toward zero, creating a simpler and more stable model.

---

## Ridge Loss Function (Matrix Form)

Prediction:

ŷ = XW

Loss Function:

L = (Y - XW)ᵀ(Y - XW) + λWᵀW

Where:

- X → Feature Matrix
- W → Weight Vector
- Y → Target Vector

---

## Expanding the Loss Function

L = (Y - XW)ᵀ(Y - XW) + λWᵀW

L = (Yᵀ - WᵀXᵀ)(Y - XW) + λWᵀW

L = YᵀY - WᵀXᵀY - YᵀXW + WᵀXᵀXW + λWᵀW

Since:

WᵀXᵀY = YᵀXW

Therefore:

L = YᵀY - 2(WᵀXᵀY) + WᵀXᵀXW + λWᵀW

---

## Derivative of the Loss Function

Differentiate w.r.t W

Term 1:

∂(YᵀY)/∂W = 0

Term 2:

∂[-2(WᵀXᵀY)]/∂W = -2XᵀY

Term 3:

∂(WᵀXᵀXW)/∂W = 2XᵀXW

Term 4:

∂(λWᵀW)/∂W = 2λW

Combining all terms:

∂L/∂W = -2XᵀY + 2XᵀXW + 2λW

---

## Why Use Gradient Descent?

The OLS solution:

W = (XᵀX + λI)⁻¹XᵀY

directly computes the optimal weights.

However:

- Matrix inversion is expensive.
- Memory usage increases with features.
- Not suitable for very large datasets.

Instead, Gradient Descent finds the solution iteratively.

---

## Gradient Descent Intuition

Gradient tells us:

"Which direction increases the loss the fastest?"

To minimize loss, move in the opposite direction.

General Update Rule:

New Parameter = Old Parameter - Learning Rate × Gradient

---

## Ridge Gradient Descent Formula

Gradient:

∂L/∂W = -2XᵀY + 2XᵀXW + 2λW

Update Equation:

W = W - η(∂L/∂W)

Substituting:

W = W - η[-2XᵀY + 2XᵀXW + 2λW]

Where:

- η → Learning Rate
- λ → Regularization Parameter

---

## Sample-wise Form Used in Code

Prediction:

ŷ = XW + b

Error:

(y - ŷ)

Bias Gradient:

b_derivative = (-2/n) * Σ(y - ŷ)

Weight Gradient:

w_derivative = (-2/n) * Xᵀ(y - ŷ) + 2λW

Notice:

2λW

is the additional Ridge penalty term.

This term continuously pulls weights toward zero.

---

## Parameter Updates

Bias Update:

b = b - lr * b_derivative

Weight Update:

W = W - lr * w_derivative

These updates are repeated for multiple epochs until convergence.

---

## Complete Learning Flow

Step 1:
Initialize random weights and bias.

Step 2:
Predict values using:

ŷ = XW + b

Step 3:
Calculate prediction error:

(y - ŷ)

Step 4:
Compute gradients.

Step 5:
Update weights and bias.

Step 6:
Repeat for all epochs.

Step 7:
Reach the minimum loss and obtain the optimal Ridge Regression model.

---

## Final Intuition

Linear Regression:

Loss = Error

Ridge Regression:

Loss = Error + Regularization

Gradient Descent:

Start with random weights → Calculate Error → Calculate Gradient → Update Weights → Repeat → Reach Optimal Solution

````python
import numpy as np  

from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor ,Ridge
from sklearn.metrics import r2_score
from sklearn.datasets import load_diabetes 
````

````python
x,y = load_diabetes(return_X_y=True)
````

````python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=4)
````

### SGDRegressor with L2 penalty

````python
reg = SGDRegressor(
    penalty='l2',                # Ridge (L2) regularization
    max_iter=500,                # Maximum number of epochs/iterations
    eta0=0.1,                    # Initial learning rat
    learning_rate='constant',    # Keep learning rate fixed throughout training
    alpha=0.001)                  # Regularization strength (λ)
````

````python
reg.fit(x_train,y_train)

y_pred = reg.predict(x_test)

print('r2 score:',r2_score(y_test,y_pred))
print('m:',reg.coef_)
print('b:',reg.intercept_)
````

### Ridge Regression (regularization)

````python
rid = Ridge(
    alpha=0.001,        # Regularization strength (λ)
    max_iter=500,       # Maximum iterations for iterative solvers
    solver='sparse_cg' ) # Conjugate Gradient solver (good for large/sparse data)
````

````python
'''
| Solver        | When to Use                                                           |
| ------------- | --------------------------------------------------------------------- |
| `'auto'`      | Default choice; sklearn automatically selects the best solver         |
| `'svd'`       | Small datasets, most numerically stable, slower                       |
| `'cholesky'`  | Small to medium dense datasets, very fast                             |
| `'lsqr'`      | Large datasets, iterative approach                                    |
| `'sparse_cg'` | Large sparse datasets (many zeros)                                    |
| `'sag'`       | Large datasets with many samples                                      |
| `'saga'`      | Very large datasets, supports sparse data and advanced regularization |
| `'lbfgs'`     | Only when `positive=True` is used                                     |
'''
````

````python
rid.fit(x_train,y_train)

y_pred2 = rid.predict(x_test)

print('r2 score:',r2_score(y_test,y_pred2))
print('m:',rid.coef_)
print('b:',rid.intercept_)
````

### Code from Scratch 

````python
class MeraRidgeGD: 

    def __init__(self,epochs,learning_rate,alpha):
        self.epochs = epochs 
        self.learning_rate = learning_rate 
        self.alpha = alpha 
        self.m = None 
        self.b = None 

    def fit(self,x_train,y_train):             # ------- Notes 
        self.m = np.ones(x_train.shape[1])
        self.b = 0 
        w = np.insert(self.m,0,self.b)

        x_train = np.insert(x_train,0,1,axis=1)

        for i in range(self.epochs): 
            w_deri = np.dot(x_train.T,x_train).dot(w) - np.dot(x_train.T,y_train) + self.alpha*w 
            w = w - self.learning_rate*w_deri

        self.m = w[1:]
        self.b = w[0]
            
    def predict(self,x_test):
        return np.dot(x_test,self.m) + self.b
````

### Note A — Weight Initialization

```python
self.m = np.ones(x_train.shape[1])
self.b = 0
```

We start Gradient Descent with initial values for the model parameters.

- `self.m` → initializes all coefficients (weights) to 1.
- `self.b` → initializes the intercept (bias) to 0.

These are only starting points. Gradient Descent will iteratively update them to reach the optimal values.

---

### Note B — Combining Bias and Weights

```python
w = np.insert(self.m,0,self.b)
```

The bias and coefficients are merged into a single parameter vector:

W = [b, w₁, w₂, w₃, ...]

This allows us to perform all calculations using matrix operations instead of handling the intercept separately.

Mathematical Form:

ŷ = XW

---

### Note C — Adding the Intercept Column

```python
x_train = np.insert(x_train,0,1,axis=1)
```

A column of 1's is inserted as the first column of the feature matrix.

Why?

Without this column:

ŷ = XW + b

With the column of 1's:

ŷ = XW

The intercept becomes part of the weight vector and can be learned together with all other coefficients.

Example:

Before:

| Age | Salary |
|------|------|
| 20 | 30000 |
| 25 | 45000 |

After:

| 1 | Age | Salary |
|---|---|---|
| 1 | 20 | 30000 |
| 1 | 25 | 45000 |

---

### Note D — Computing the Ridge Gradient

```python
w_deri = np.dot(x_train.T,x_train).dot(w) - np.dot(x_train.T,y_train) + self.alpha*w

```

This calculates the gradient of the Ridge Regression loss function.

Mathematical Formula:

∂L/∂W = XᵀXW − XᵀY + λW

Where:

- XᵀXW → error gradient
- XᵀY → actual target information
- λW → Ridge penalty term

The Ridge penalty shrinks large coefficients and helps reduce overfitting.

---

### Note E — Gradient Descent Update Rule

```python
w = w - self.learning_rate * w_deri
```

Weights are updated in the opposite direction of the gradient.

General Formula:

New Parameter = Old Parameter − Learning Rate × Gradient

or

W = W − η(∂L/∂W)

Where:

- η (eta) = Learning Rate
- ∂L/∂W = Gradient

This step gradually moves the model toward the minimum loss.

---

### Note F — Extracting Final Parameters

```python
self.m = w[1:]
self.b = w[0]
```

After training is complete:

- `w[0]` becomes the intercept (bias)
- `w[1:]` becomes the coefficients (weights)

Final Prediction Equation:

ŷ = XW + b

These learned values are then used to make predictions on unseen data.

````python
mera = MeraRidgeGD(epochs=500 , learning_rate= 0.005 , alpha=0.001)
````

````python
mera.fit(x_train,y_train)

y_pred3 = mera.predict(x_test)

print('r2 score:',r2_score(y_test,y_pred3))
print('m:',mera.m)
print('b:',mera.b)
````
