# Lasso Regression (L1 Regularization)

## Main Intuition

The most important idea behind **Lasso Regression** is that it not only reduces overfitting but also performs **automatic feature selection**.

As the regularization parameter (**λ / alpha**) increases:

* The coefficient values begin to shrink.
* Less important features shrink faster.
* Some coefficients become **exactly zero**.
* Features whose coefficients become zero are effectively removed from the model.

This is the key difference between **Ridge Regression** and **Lasso Regression**:

| Ridge (L2)                       | Lasso (L1)                                       |
| -------------------------------- | ------------------------------------------------ |
| Coefficients shrink towards zero | Coefficients can become exactly zero             |
| Keeps all features               | Can remove features                              |
| Reduces overfitting              | Reduces overfitting + performs feature selection |
| Does not reduce dimensionality   | Reduces dimensionality                           |

Because Lasso can eliminate unnecessary features, the final model often becomes:

* Simpler
* Easier to interpret
* Faster to train and predict
* Less prone to overfitting

In simple terms:

> If a feature is not contributing enough to the prediction, Lasso can completely remove it from the model.

---

# Key Concepts

### What is Lasso Regression?

Lasso Regression adds an **L1 penalty** (absolute value of coefficients) to the Mean Squared Error (MSE) loss function.

Loss Function:

L = MSE + λ Σ|w|

The penalty discourages large coefficient values and helps control model complexity.

---

### Feature Selection

One of the biggest advantages of Lasso Regression is its ability to perform **feature selection**.

As λ increases:

* Important features retain larger coefficients.
* Less useful features shrink.
* Some coefficients become exactly zero.

This automatically removes irrelevant or redundant columns from the model.

---

### Effect of Lambda (λ)

The regularization parameter controls the strength of the penalty.

* Small λ → Model behaves similarly to Linear Regression.
* Medium λ → Balanced regularization.
* Large λ → More coefficients shrink to zero.

Choosing the correct λ is important because excessive regularization can lead to underfitting.

---

### Bias-Variance Tradeoff

Just like Ridge Regression, Lasso affects the bias-variance tradeoff.

As λ increases:

* Bias increases
* Variance decreases

The objective is to find an optimal λ where the model generalizes well to unseen data.

---

### Effect on the Loss Function

The geometry of the L1 penalty creates corner points in the optimization space.

Unlike Ridge Regression, these corners make it possible for coefficients to become exactly zero.

This geometric property is what enables feature selection.

---

## Why Lasso is Popular

Lasso is widely used because it can:

* Reduce overfitting
* Select important features automatically
* Reduce dimensionality
* Improve model interpretability
* Simplify high-dimensional datasets

This makes it particularly useful when working with datasets containing many input features.

# (IMP interview Ques) Ridge vs. Lasso
### Question : Why can Lasso Regression make coefficients exactly zero, while Ridge Regression cannot?

### 1. Ridge Regression Formula (L2 Regularization)

The closed-form solution for the slope $m$ in Ridge regression adds the penalty to the `denominator`:

$$m = \frac{\sum (y_i - \bar{y})(x_i - \bar{x})}{\sum (x_i - \bar{x})^2 + \lambda}$$

---

### 2. Lasso Regression Formula (L1 Regularization)

Because the L1 penalty uses absolute values, the solution for $m$ depends on its sign, creating three distinct cases:

* **For $m > 0$:**

$$m = \frac{\sum (y_i - \bar{y})(x_i - \bar{x}) - \lambda}{\sum (x_i - \bar{x})^2}$$


* **For $m = 0$:**

$$m = \frac{\sum (y_i - \bar{y})(x_i - \bar{x})}{\sum (x_i - \bar{x})^2}$$


* **For $m < 0$:**

$$m = \frac{\sum (y_i - \bar{y})(x_i - \bar{x}) + \lambda}{\sum (x_i - \bar{x})^2}$$



---

## Quick Comparison: How They Behave

> **Ridge (L2):** Adds the penalty ($\lambda$) directly to the `denominator`.
> * **Effect:** Increasing $\lambda$ increases the denominator, which smoothly shrinks the coefficient $m$ closer and closer to zero. However, mathematically, it will **never reach exactly zero**.
> 
> 

> **Lasso (L1):** Subtracts or adds the penalty ($\lambda$) to the `numerator` (depending on the direction of $m$).
> * **Effect:** This mathematical shift pulls the numerator directly toward zero. If the penalty $\lambda$ is large enough, it completely overpowers the data term, forcing $m$ to be **exactly zero**. This is what creates **coefficient sparsity** and allows Lasso to act as an automatic feature selector.
> 
>

# Key Points

## 1. How are coefficients affected?

```python
from sklearn.datasets import load_diabetes

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score

from sklearn.model_selection import train_test_split

import warnings
warnings.filterwarnings('ignore')
```

```python
data = load_diabetes()

df = pd.DataFrame(data.data,columns=data.feature_names)
df['TARGET'] = data.target

df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(data.data,data.target,test_size=0.2,random_state=2)
```

```python
coefs = []
r2_scores = []

for i in [0,0.1,1,10]:
    reg = Lasso(alpha=i)
    reg.fit(X_train,y_train)
    
    coefs.append(reg.coef_.tolist())
    y_pred = reg.predict(X_test)
    r2_scores.append(r2_score(y_test,y_pred))
```

```python
plt.figure(figsize=(14,9))
plt.subplot(221)
plt.bar(data.feature_names,coefs[0])
plt.title('Alpha = 0 ,r2_score = {}'.format(round(r2_scores[0],2)))

plt.subplot(222)
plt.bar(data.feature_names,coefs[1])
plt.title('Alpha = 0.1 ,r2_score = {}'.format(round(r2_scores[1],2)))

plt.subplot(223)
plt.bar(data.feature_names,coefs[2])
plt.title('Alpha = 1 ,r2_score = {}'.format(round(r2_scores[2],2)))

plt.subplot(224)
plt.bar(data.feature_names,coefs[3])
plt.title('Alpha = 10 ,r2_score = {}'.format(round(r2_scores[3],2)))

plt.show()
```

```python
# As the alpha (lamda) increases the less relevent features(columns) becomes zero 
```

## 3. Impact on Bias and Variance

```python
m = 100
X = 5 * np.random.rand(m, 1) - 2
y = 0.7 * X ** 2 - 2 * X + 3 + np.random.randn(m, 1)

plt.scatter(X, y)
plt.show()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X.reshape(100,1),y.reshape(100),test_size=0.2,random_state=2)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=10)

X_train = poly.fit_transform(X_train)
X_test = poly.transform(X_test)
```

```python
from mlxtend.evaluate import bias_variance_decomp

alphas = np.linspace(0,30,100)

loss = []
bias = []
variance = []

for i in alphas:
    reg = Lasso(alpha=i)
    avg_expected_loss, avg_bias, avg_var = bias_variance_decomp(
        reg, X_train, y_train, X_test, y_test, 
        loss='mse',
        random_seed=123)
    loss.append(avg_expected_loss)
    bias.append(avg_bias)
    variance.append(avg_var)
```

```python
plt.plot(alphas,loss,label='loss')
plt.plot(alphas,bias,label='Bias')
plt.plot(alphas,variance,label='Variance')
plt.xlabel('Alpha')
plt.legend()
```

## 4. Effect of Regularization on Loss Function

```python
from sklearn.datasets import make_regression

X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)

plt.scatter(X,y)

from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X,y)
print(reg.coef_)
print(reg.intercept_)
```

```python
def cal_loss(m,alpha):
    return np.sum((y - m*X.ravel() + 2.29)**2) + alpha*abs(m)
```

```python
def predict(m):
    return m*X - 2.29
```

```python
m = np.linspace(-45,100,100)
plt.figure(figsize=(12,12))
for j in [0,100,500,1000,2500,3500,4500,5500]:
    loss = []
    for i in range(m.shape[0]):
        loss_i = cal_loss(m[i],j)
        loss.append(loss_i)
    plt.plot(m,loss,label='alpha = {}'.format(j))
plt.legend()
plt.xlabel('Alpha')
plt.ylabel('Loss')
plt.show()
```
