# ElasticNet Regression

## Why Was ElasticNet Created?

Before ElasticNet, we had two popular regularization techniques:

### Ridge Regression

$$
L = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda |w|^2
$$

* Works well when most features are useful.
* Handles multicollinearity effectively.
* Shrinks coefficients towards zero.
* Cannot perform feature selection because coefficients rarely become exactly zero.

---

### Lasso Regression

$$
L = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda |w|
$$

* Performs automatic feature selection.
* Can shrink coefficients exactly to zero.
* Reduces dimensionality of the dataset.

However, when multiple features are highly correlated, Lasso may keep only one feature and remove the others, even though they may contain useful information.

---

## Why Do We Need ElasticNet?

In many real-world datasets:

* Some features are important.
* Some features are irrelevant.
* Many features are highly correlated.

In such situations, we want:

* The feature selection ability of Lasso.
* The coefficient stabilization of Ridge.

ElasticNet was created to combine the strengths of both techniques.

> **Rule of Thumb:** If a dataset contains many features and we are unsure which features are relevant or irrelevant, ElasticNet is often a good choice because it combines the advantages of both Ridge and Lasso Regression.

---

## ElasticNet Loss Function

$$
L = \sum (y_i - \hat{y}_i)^2 + a|w|^2 + b|w|
$$

Where:

* $\sum (y_i - \hat{y}_i)^2$ → Residual Sum of Squares (RSS)
* $a|w|^2$ → Ridge (L2) penalty
* $b|w|$ → Lasso (L1) penalty

ElasticNet therefore applies both penalties simultaneously.

---

## Hyperparameters

ElasticNet introduces two important hyperparameters.

### Lambda

$$
lambda = a + b
$$

Controls the overall strength of regularization.

* Larger $\lambda$ → More regularization
* Smaller $\lambda$ → Less regularization

---

### L1 Ratio

$$
l1-ratio = \frac{a}{a+b}
$$

Controls how much influence comes from Ridge and Lasso.

* Closer to 0 → More Ridge-like behavior
* Closer to 1 → More Lasso-like behavior
* Between 0 and 1 → Combination of both

---

## Main Intuition

ElasticNet is essentially:

$$
\text{ElasticNet} = \text{Ridge} + \text{Lasso}
$$

It attempts to:

* Keep useful correlated features (Ridge behavior).
* Remove irrelevant features (Lasso behavior).
* Reduce overfitting.
* Improve model generalization.

This makes ElasticNet particularly useful for datasets with a large number of features and multicollinearity.


````python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
````

````python
x,y = load_diabetes(return_X_y=True)
````

````python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
````

````python
# Linear Regression
reg = LinearRegression()
reg.fit(x_train,y_train)
y_pred = reg.predict(x_test)
r2_score(y_test,y_pred)
````

````python
# Ridge 
rid = Ridge(alpha=0.1)
rid.fit(x_train,y_train)
y_pred = rid.predict(x_test)
r2_score(y_test,y_pred)
````

````python
# Lasso
las = Lasso(alpha=0.01)
las.fit(x_train,y_train)
y_pred = las.predict(x_test)
r2_score(y_test,y_pred)
````

````python
# ElasticNet
ent = ElasticNet(alpha=0.005,l1_ratio=0.9)
ent.fit(x_train,y_train)
y_pred = ent.predict(x_test)
r2_score(y_test,y_pred)
````

````python
# elastic out performces Ridge & Lasso 
````

## Implementation in Scikit-Learn

ElasticNet can be implemented in Scikit-Learn using two approaches:

1. **ElasticNet**

   * Dedicated implementation of ElasticNet Regression.
   * Recommended for most use cases.

```python
from sklearn.linear_model import ElasticNet

reg = ElasticNet(alpha=0.1, l1_ratio=0.5)
```
---
2. **SGDRegressor**

   * Set `penalty='elasticnet'`.
   * Uses Gradient Descent optimization and is useful for large datasets.

```python
from sklearn.linear_model import SGDRegressor

reg = SGDRegressor(
    penalty='elasticnet',
    alpha=0.1,
    l1_ratio=0.5
)
```
