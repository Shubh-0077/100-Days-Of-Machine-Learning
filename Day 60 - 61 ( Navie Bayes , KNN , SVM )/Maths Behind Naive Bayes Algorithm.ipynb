## 1. Problem Formulation
* **Goal:** To build a supervised machine learning classification model using probability.
* **Input Features:** A data point represented as an $n$-dimensional feature vector:
  $$\mathbf{X} = [X_1, X_2, X_3, \dots, X_n]$$
* **Target Variable:** The model needs to classify $\mathbf{X}$ into one of $k$ distinct classes:
  $$\text{Classes} = \{C_1, C_2, C_3, \dots, C_k\}$$
* **Core Objective:** Find the probability of each class given the input features, and assign the data point to the class with the highest probability:
  $$P(C_k \mid X_1, X_2, \dots, X_n)$$

---

## 2. Applying Bayes' Theorem
To compute the conditional probability, we apply **Bayes' Theorem**:

$$P(C_k \mid X_1, X_2, \dots, X_n) = \frac{P(X_1, X_2, \dots, X_n \mid C_k) \cdot P(C_k)}{P(X_1, X_2, \dots, X_n)}$$

### Breaking Down the Components:
1. **Posterior Probability ($P(C_k \mid \mathbf{X})$):** The probability that the data point belongs to class $C_k$ given its features. This is what we want to calculate.
2. **Prior Probability ($P(C_k)$):** The baseline probability of a class occurring in the training data (e.g., total count of class $C_k$ divided by total dataset size).
3. **Likelihood ($P(\mathbf{X} \mid C_k)$):** The probability of observing this specific combination of features given that the class is $C_k$.
4. **Evidence ($P(\mathbf{X})$):** The total probability of the features occurring. 

### Simplifying the Equation:
* Notice that the **Evidence ($P(X_1, X_2, \dots, X_n)$)** in the denominator is identical for every single class $C_k$. 
* Since it acts as a constant scaling factor and does not change the relative ranking of the classes, we can drop the denominator for optimization purposes:
  $$P(C_k \mid X_1, X_2, \dots, X_n) \propto P(X_1, X_2, \dots, X_n \mid C_k) \cdot P(C_k)$$
  *(Where $\propto$ means "proportional to")*

---

## 3. The Joint Likelihood Bottleneck
* **The Problem:** Calculating the term $P(X_1, X_2, \dots, X_n \mid C_k)$ directly from a dataset is functionally impossible for large numbers of features. The exact combination of all features occurring simultaneously under one class is incredibly rare or won't exist in our training data.
* To expand this joint probability term mathematically, we must use the **Chain Rule of Conditional Probability**:

$$P(X_1, X_2, \dots, X_n \mid C_k) = P(X_1 \mid X_2, \dots, X_n, C_k) \cdot P(X_2 \mid X_3, \dots, X_n, C_k) \dots P(X_n \mid C_k)$$

* **Why this fails practically:** This expands into highly nested, deeply dependent probabilities (e.g., the probability of $X_1$ depends on the state of all other features $X_2$ through $X_n$ as well as class $C_k$). This remains computationally intractable to calculate.

---

## 4. The "Naive" Assumption: Conditional Independence
To overcome the joint likelihood problem, the algorithm makes a major simplifying assumption: **All features are conditionally independent of each other given the class label $C_k$.**

### Mathematical Impact of Independence:
If features are completely independent of one another given the class, knowing the value of one feature gives zero information about any other feature. Thus, the complex conditional dependencies disappear:
* $P(X_1 \mid X_2, \dots, X_n, C_k) \approx P(X_1 \mid C_k)$
* $P(X_2 \mid X_3, \dots, X_n, C_k) \approx P(X_2 \mid C_k)$
* $\dots$ and so forth.

Therefore, the highly complex joint likelihood term collapses simply into a **product of individual feature likelihoods**:

$$P(X_1, X_2, \dots, X_n \mid C_k) = P(X_1 \mid C_k) \cdot P(X_2 \mid C_k) \dots P(X_n \mid C_k)$$

Using product notation ($\prod$), we write it compactly as:
$$P(X_1, X_2, \dots, X_n \mid C_k) = \prod_{i=1}^{n} P(X_i \mid C_k)$$

---

## 5. Final Derivation and Decision Rule
Substituting the simplified product back into our original proportional equation gives us the finalized formula for the Naive Bayes Classifier:

$$P(C_k \mid X_1, X_2, \dots, X_n) \propto P(C_k) \cdot \prod_{i=1}^{n} P(X_i \mid C_k)$$

### The MAP Decision Rule:
To make a final prediction ($\hat{y}$), the model calculates this value for all $k$ classes and chooses the class that yields the maximum value. This is known as the **Maximum A Posteriori (MAP)** decision rule:

$$\hat{y} = \arg\max_{k \in \{1, \dots, k\}} \left( P(C_k) \cdot \prod_{i=1}^{n} P(X_i \mid C_k) \right)$$

---

## 6. Key Teacher Takeaways
* **Why it's called "Naive":** Real-world features are almost never completely independent of each other (e.g., words like "Mumbai" and "Chennai" are highly related when discussing cricket or regions). Ignoring these dependencies is inherently flawed or "naive".
* **Why we use it anyway:** Despite the unrealistic assumption, it transforms an impossible mathematical problem into a series of incredibly simple, individual frequency lookups ($P(X_i \mid C_k)$). This makes Naive Bayes incredibly fast, highly scalable to high-dimensional datasets (like text classification), and surprisingly accurate in practice.

# Naive Bayes for Numerical Data (Gaussian Naive Bayes)
vid : https://youtu.be/TCgK2nBJx9o?si=smbMuXvmDAvbEwjR 

## Why Do We Need Gaussian Naive Bayes?

Traditional Naive Bayes works well with categorical data, where probabilities are calculated using frequency counts.

For example:

P(Outlook = Sunny | Play = Yes)

can be computed by counting occurrences in the dataset.

However, this approach fails when dealing with continuous numerical features such as:

- Height
- Weight
- Age
- Salary

### The Problem

Suppose we receive a new data point:

Height = 175 cm

Weight = 80 kg

If we try to calculate:

P(Height = 175 | Male)

using frequency counts, the probability will likely be 0 because an exact value of 175 cm may never appear in the training dataset.

Since Naive Bayes multiplies probabilities together, even one zero probability causes the entire prediction to become zero.

To solve this problem, we use **Gaussian Naive Bayes**.

---

# Sample Dataset

| Person | Height (cm) | Weight (kg) | Gender |
|----------|------------|------------|---------|
| 1 | 180 | 82 | Male |
| 2 | 173 | 75 | Male |
| 3 | 178 | 78 | Male |
| 4 | 162 | 53 | Female |
| 5 | 165 | 60 | Female |
| 6 | 158 | 50 | Female |

# Query Example

Given:

Height = 175 cm

Weight = 80 kg

Predict:

Male or Female

Using Naive Bayes:

$$
P(Male \mid Height=175, Weight=80)
\propto
P(Male)
\times
P(Height=175 \mid Male)
\times
P(Weight=80 \mid Male)
$$

---

# Gaussian Assumption

Gaussian Naive Bayes assumes that each numerical feature follows a Normal Distribution (Bell Curve) within every class.

Instead of counting frequencies, we model each feature using:

- Mean ($\mu$)
- Standard Deviation ($\sigma$)

---

# Step 1: Calculate Mean and Standard Deviation

### Mean

$$
\mu = \frac{\sum x_i}{n}
$$

### Standard Deviation

$$
\sigma = \sqrt{\frac{\sum (x_i-\mu)^2}{n}}
$$

Example for Male class:

Height values:

180, 173, 178

Calculate:

$$
\mu_{height,male}
$$

$$
\sigma_{height,male}
$$

Weight values:

82, 75, 78

Calculate:

$$
\mu_{weight,male}
$$

$$
\sigma_{weight,male}
$$

---

# Step 2: Use Gaussian Probability Density Function (PDF)

$$
f(x \mid \mu,\sigma)
=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- $x$ = feature value
- $\mu$ = mean
- $\sigma$ = standard deviation

This function returns a probability density score instead of a simple frequency count.

---

# Classification Pipeline

## 1. Train the Model

For every feature and class:

- Calculate Mean ($\mu$)
- Calculate Standard Deviation ($\sigma$)

## 2. Compute Male Score

### Height Likelihood

$$
P(Height=175 \mid Male)
=
f(175 \mid \mu_{height,male}, \sigma_{height,male})
$$

### Weight Likelihood

$$
P(Weight=80 \mid Male)
=
f(80 \mid \mu_{weight,male}, \sigma_{weight,male})
$$

### Prior Probability

$$
P(Male)
=
\frac{3}{6}
=
0.5
$$

### Final Male Score

$$
Score_{Male}
=
P(Male)
\times
P(Height=175 \mid Male)
\times
P(Weight=80 \mid Male)
$$

## 3. Compute Female Score

$$
Score_{Female}
=
P(Female)
\times
P(Height=175 \mid Female)
\times
P(Weight=80 \mid Female)
$$

## 4. Final Prediction

If

$$
Score_{Male} > Score_{Female}
$$

Predict Male.

Otherwise predict Female.

# Other Variants of Naive Bayes

## Gaussian Naive Bayes

Used when features are continuous numerical values and approximately follow a Normal Distribution.

Examples:

- Height
- Weight
- Age
- Salary

## Bernoulli Naive Bayes

Used when features are binary.

Examples:

- 0 or 1
- Yes or No
- True or False

## Multinomial Naive Bayes

Used when features represent counts.

Examples:

- Word frequencies
- Token counts
- Document classification

---

# Main Intuition

Traditional Naive Bayes calculates probabilities using frequency counts.

For continuous numerical data, exact values rarely repeat, making frequency-based probabilities unreliable.

Gaussian Naive Bayes solves this problem by assuming that each feature follows a Normal Distribution and uses the Gaussian Probability Density Function (PDF) to calculate likelihoods.

This allows Naive Bayes to work effectively with continuous numerical features such as height, weight, age, income, and many other real-world measurements.
