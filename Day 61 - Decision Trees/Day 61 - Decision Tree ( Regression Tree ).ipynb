# 1. Introduction: Classification vs Regression Trees

Decision Trees are not limited to classification problems. They can also be used for **Regression Problems**, where the target variable is continuous.

## Classification Tree

Target variable:

```text id="6jjlq7"
Yes / No
Spam / Not Spam
Pass / Fail
```

Prediction:

```text id="9j3p2r"
Discrete Classes
```

---

## Regression Tree

Target variable:

```text id="xej7mi"
Marks
House Price
Salary
Temperature
```

Prediction:

```text id="fz6wz4"
Continuous Numerical Values
```

---

# Linear Regression vs Regression Trees

## Linear Regression

Works well when:

```text id="pgt0g4"
Input and Output have a linear relationship.
```

Equation:

$$
y = mx + c
$$

---

## Regression Trees

Used when:

```text id="s4ad7a"
The relationship is highly non-linear.
```

A single straight line cannot capture the pattern.

---

# 2. Geometric & Conceptual Intuition

## Example

Input:

```text id="sv6vkh"
Hours Studied
```

Output:

```text id="8ps7gx"
Exam Marks
```

---

# Case 1: Linear Relationship

As study hours increase:

```text id="4yx7v7"
Marks increase linearly.
```

A straight line fits perfectly.

---

# Case 2: Non-Linear Relationship

Suppose:

```text id="pc32dl"
0-3 hours   → Low Marks
3-6 hours   → High Marks
>6 hours    → Low Marks
```

This relationship is not linear.

Linear Regression performs poorly.

Regression Trees solve this by dividing the feature space into multiple regions.

---

# Regression Tree Structure

```text id="c2g0n0"
                Is Hours < 3?
                     /      \
                   Yes      No
                   /          \
            Mean of R1      Is Hours < 6?
                               /      \
                             Yes      No
                             /          \
                      Mean of R2      Mean of R3
```

---

# Prediction Mechanism

Suppose:

```text id="tsjlwm"
Hours Studied = 5
```

The sample travels through the tree and lands inside Region 2.

The prediction is:

```text id="4z3qeh"
Mean Marks of Region 2
```

---

# Main Intuition

Regression Trees do not fit a line.

Instead, they:

1. Divide the feature space.
2. Compute the average target value in each region.
3. Predict using that average.

---

# 3. Mathematical Mechanics: Choosing Split Points

Classification Trees use:

* Entropy
* Gini Impurity

Regression Trees use:

* Variance Reduction
* Mean Squared Error (MSE)
* Sum of Squared Residuals (SSR)

---

# Step 1: Sort the Feature

Arrange the feature values in ascending order.

---

# Step 2: Generate Candidate Splits

Try every possible threshold.

Example:

```text id="hnbwzr"
Hours ≤ 1
Hours ≤ 2
Hours ≤ 3
Hours ≤ 4
```

---

# Step 3: Partition the Data

Left Group:

$$
X \le X_{split}
$$

Right Group:

$$
X > X_{split}
$$

---

# Step 4: Calculate Regional Means

Left Mean:

$$
\bar{y}_L
$$

Right Mean:

$$
\bar{y}_R
$$

---

# Step 5: Compute SSR

$$
SSR =
\sum_{i \in Left}(y_i-\bar{y}*L)^2
+
\sum*{j \in Right}(y_j-\bar{y}_R)^2
$$

---

# Step 6: Select the Best Split

Choose the split having the:

```text id="yj4wnh"
Minimum SSR
```

because minimum SSR means:

```text id="vgu0c6"
Minimum Error
Maximum Homogeneity
```

---

# Main Objective of Regression Trees

$$
\text{Minimize SSR}
$$

or

$$
\text{Minimize MSE}
$$

---

# 4. Multi-Dimensional Expansion

Suppose:

Features:

```text id="11r5ui"
X1 = Hours Studied

```

Target:

```text id="9j6tly"
Y = Exam Marks
```

---

# How Splitting Works

## Step 1

Find the best split on:

```text id="m2h0hk"
Hours Studied
```

Compute its minimum SSR.

---

## Step 2

Find the best split on:

```text id="yyqce5"
CGPA
```

Compute its minimum SSR.

---

## Step 3

Compare both SSR values.

Whichever feature produces:

```text id="h8djlwm"
Lowest SSR
```

wins the split.

---

# Geometric Intuition

Regression Trees partition the feature space into:

* Rectangles
* Cuboids
* Hyper-rectangles

depending upon the number of dimensions.

# 5. Hyperparameters of Regression Trees

---

# criterion

Determines how split quality is measured.

Options:

```python
criterion='squared_error'
criterion='absolute_error'
criterion='friedman_mse'
```

---

## squared_error

Default option.

Uses:

```text id="llz0yc"
MSE
Variance Reduction
```

---

## absolute_error

Uses:

```text id="6pfbzc"
Mean Absolute Error
```

---

# splitter

Options:

```python
splitter='best'
splitter='random'
```

---

## best

Checks every split.

More accurate but can overfit.

---

## random

Checks random splits.

Faster and reduces overfitting.

---

# max_depth

Controls tree depth.

```python
max_depth=None
```

means:

```text id="0h55os"
Grow until pure leaves.
```

---

## Small Value

```python
max_depth=1
```

Produces:

* Underfitting
* High Bias

---

## Large Value

Produces:

* Overfitting
* High Variance

---

# min_samples_split

Minimum samples required before splitting.

Higher value:

* Simpler Tree
* Less Overfitting

Lower value:

* Deeper Tree
* More Overfitting

---

# min_samples_leaf

Minimum samples allowed inside a leaf node.

Higher value:

* Larger regions
* Less overfitting

Lower value:

* Tiny regions
* More overfitting

---

# min_impurity_decrease

A split occurs only if:

```text id="0nvlgo"
Error Reduction > Threshold
```

Higher values:

* Early stopping
* Simpler trees

---

# 6. Python Implementation

# Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth':[3,5,7,10],
    'criterion':['squared_error','friedman_mse'],
    'max_features':[0.5,1.0],
    'min_samples_split':[10,25,50]
}

grid = GridSearchCV(
    DecisionTreeRegressor(),
    param_grid,
    cv=5
)

grid.fit(X_train,y_train)

print(grid.best_params_)
```

---

# Advantages of Regression Trees

1. Captures Non-Linear Relationships.
2. Easy to Interpret.
3. No Feature Scaling Required.
4. Handles Feature Interactions Automatically.
5. Performs Feature Selection.

---

# Disadvantages of Regression Trees

1. Prone to Overfitting.
2. Unstable to Small Data Changes.
3. Produces Piecewise Constant Predictions.
4. Usually Less Accurate than Ensemble Methods.

---

# Summary Table

| Concept                  | Regression Tree   |
| ------------------------ | ----------------- |
| Target Variable          | Continuous        |
| Prediction               | Mean of Region    |
| Split Criterion          | Minimum SSR / MSE |
| Boundary Shape           | Axis Parallel     |
| Handles Non-Linearity    | Yes               |
| Feature Scaling Required | No                |
| Major Problem            | Overfitting       |

---

# Hyperparameter Cheat Sheet

| Hyperparameter          | Reduce Overfitting | Reduce Underfitting |
| ----------------------- | ------------------ | ------------------- |
| `max_depth`             | Decrease           | Increase            |
| `min_samples_split`     | Increase           | Decrease            |
| `min_samples_leaf`      | Increase           | Decrease            |
| `splitter`              | Use `'random'`     | Use `'best'`        |
| `min_impurity_decrease` | Increase           | Decrease            |

---

# Final Intuition

Regression Trees do not try to fit a single equation like:

$$
y = mx + c
$$

Instead, they:

1. Divide the feature space into regions.
2. Find the average target value inside each region.
3. Predict using that average.

Their entire learning process revolves around one goal:

$$
\boxed{\text{Find splits that minimize SSR and MSE}}
$$


````python
import numpy as np 
import pandas as pd 
from pandas_datareader import data

from sklearn.model_selection import train_test_split 
from sklearn.metrics import r2_score 

from sklearn.tree import DecisionTreeRegressor

from sklearn.datasets import fetch_california_housing 
from sklearn.model_selection import GridSearchCV 
````

````python

# Load dataset
housing = fetch_california_housing()

# Convert features into DataFrame
df = pd.DataFrame(
    housing.data,
    columns=housing.feature_names
)

# Add target column
df['MedHouseValue'] = housing.target

# Display first 5 rows
df.head()
````

````python
x = df.iloc[:,0:8]
y = df.iloc[:,-1]
````

````python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2,random_state=42)
````

````python
rt = DecisionTreeRegressor(criterion = 'squared_error', max_depth=5)
````

````python
rt.fit(x_train,y_train)
````

````python
y_pred = rt.predict(x_test)
````

````python
r2_score(y_test,y_pred)
````

### Hyperparameters

````python
param_grid = {
    'max_depth': [2, 4, 8, 10, None],
    'criterion': ['squared_error', 'absolute_error'],
    'max_features': [0.25, 0.5, 1.0],
    'min_samples_split': [2, 5, 10, 20]
}
````

````python
reg = GridSearchCV(DecisionTreeRegressor(),param_grid=param_grid)
     

reg.fit(x_train,y_train)
````

````python
# Accuracy  
reg.best_score_
````

````python
reg.best_params_
````

### Feature Importance 

````python
# This will help us to find the best collerated cols to target variables  

for importance, name in sorted(zip(rt.feature_importances_, x_train.columns),reverse=True):
  print (name, importance)
````
