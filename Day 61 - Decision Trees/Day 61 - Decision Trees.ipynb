# 1. Introduction to Decision Trees

## What is a Decision Tree?

A Decision Tree is one of the most intuitive Machine Learning algorithms.

Programmatically, it is simply a nested structure of:

```text
IF → ELSE → IF → ELSE
```

conditions.

Decision Trees make predictions by asking a sequence of questions about the data.

---

## Tree Structure

### Root Node

The topmost node where the first split occurs.

### Internal Nodes

Intermediate decision nodes that further split the data.

### Leaf Nodes

The final prediction nodes.

---

## Example: App Recommendation Dataset

| Gender | Occupation | App      |
| ------ | ---------- | -------- |
| Male   | Student    | PUBG     |
| Female | Student    | PUBG     |
| Male   | Working    | WhatsApp |
| Female | Working    | Greta    |

### Logic

```text
If Occupation == Student:
    Recommend PUBG

Else:
    If Gender == Female:
        Recommend Greta
    Else:
        Recommend WhatsApp
```

---

# 2. Geometric Intuition

When working with numerical features, Decision Trees partition the feature space into smaller regions.

## Axis-Parallel Splits

Decision Trees always create boundaries that are:

* Vertical lines
* Horizontal lines
* Hyperplanes parallel to the coordinate axes

They never create diagonal boundaries like Logistic Regression or SVM.

---

## Hyper-Cuboids

The recursive splits divide the feature space into boxes.

### In 2D

Rectangles

### In 3D

Cuboids

### In Higher Dimensions

Hyper-cuboids

---

# Main Intuition

Decision Trees keep asking questions until each region becomes sufficiently pure.

---

# 3. General Recursive Algorithm

1. Start with the entire dataset.
2. Find the best feature to split the data.
3. Create child datasets.
4. Repeat the process recursively.
5. Stop when:

   * Pure leaf nodes are obtained.
   * Maximum depth is reached.
   * Minimum samples condition is satisfied.

---

# 4. Measuring Impurity: Entropy

## What is Entropy?

Entropy measures:

* Impurity
* Disorder
* Uncertainty

Main idea:

> More Knowledge → Less Uncertainty → Lower Entropy

---

## Entropy Formula

$$
Entropy(S) = -p_{+}\log_2(p_{+}) - p_{-}\log_2(p_{-})
$$

where:

* $p_{+}$ = probability of positive class
* $p_{-}$ = probability of negative class

---

## Properties of Entropy

### Pure Node

$$
Entropy = 0
$$

Example:

```text
Yes Yes Yes Yes
```

or

```text
No No No No
```

---

### Maximum Impurity

$$
Entropy = 1
$$

Example:

```text
Yes Yes No No
```

---

### Entropy Range

$$
0 \leq Entropy \leq 1
$$

For multi-class classification, entropy can exceed 1.

---

# 5. Information Gain (IG)

## What is Information Gain?

Information Gain measures how much uncertainty decreases after a split.

The goal is:

> Maximize Information Gain.

---

## Information Gain Formula

$$
IG = Entropy(Parent)
-
\sum_{i=1}^{k}
\left(
\frac{|D_i|}{|D|}
\times
Entropy(D_i)
\right)
$$
--
where:

* $|D_i|$ = number of samples in child node
* $|D|$ = number of samples in parent node

---

## Steps to Calculate Information Gain

1. Calculate Parent Entropy.
2. Split the dataset.
3. Calculate Child Entropies.
4. Compute Weighted Child Entropy.
5. Subtract from Parent Entropy.

The feature with the highest Information Gain becomes the next split.

---

# 6. Measuring Impurity: Gini Impurity

Gini Impurity is the default criterion used in:

```python
DecisionTreeClassifier()
```

---

## Gini Formula

$$
Gini = 1 - \sum_{i=1}^{C} p_i^2
$$

For binary classification:

$$
Gini = 1 - p_{yes}^2 - p_{no}^2
$$

---

## Properties of Gini

### Pure Node

$$
Gini = 0
$$

---

### Maximum Impurity

$$
Gini = 0.5
$$

Example:

```text
50% Yes
50% No
```

---

### Gini Range

$$
0 \leq Gini \leq 0.5
$$

---

# Gini vs Entropy

## Gini

* Faster computation
* Uses squares only
* Default in Scikit-Learn

## Entropy

* Uses logarithms
* Slightly slower
* Sometimes produces cleaner splits

---

# 7. Handling Continuous Features

For numerical features such as:

* Age
* Salary
* Rating
* Petal Length

the algorithm cannot directly split by categories.

---

## Workflow

### Step 1

Sort the feature.

### Step 2

Try every possible threshold.

Example:

```text
Age > 18
Age > 22
Age > 30
```

### Step 3

Compute Information Gain (or Gini Reduction).

### Step 4

Choose the threshold that gives the maximum Information Gain.

---

# Training Complexity

Training can become computationally expensive because many candidate splits need to be evaluated.

However, once the tree is built:

## Prediction Complexity

$$
O(\log N)
$$

which makes Decision Trees very fast during inference.

---

# Advantages of Decision Trees

### 1. Easy to Understand

Highly interpretable and explainable.

### 2. Minimal Preprocessing

No need for:

* Feature Scaling
* Normalization
* Standardization

### 3. Fast Prediction

$$
O(\log N)
$$

### 4. Supports Both Tasks

* Classification
* Regression

This framework is called:

```text
CART
(Classification And Regression Trees)
```

---

# Disadvantages of Decision Trees

### 1. Prone to Overfitting

Deep trees can memorize the training data.

### 2. Unstable

Small changes in data can lead to completely different trees.

### 3. Sensitive to Imbalanced Data

The tree may become biased toward the majority class.

### 4. Greedy Nature

A locally optimal split does not always lead to a globally optimal tree.

---

# Main Intuition

A Decision Tree repeatedly asks:

> Which question reduces uncertainty the most?

The algorithm keeps splitting the data until the resulting regions become sufficiently pure.

This entire process is driven by:

* Entropy
* Information Gain
* Gini Impurity

which act as measures of impurity and help the tree decide the best split at every node.


# Decision Trees: Hyperparameters, Overfitting & Underfitting

---

# 1. The Core Problem: Overfitting in Decision Trees

Decision Trees are highly **greedy algorithms**.

If no restrictions are applied, the tree will continue splitting until:

* Every leaf becomes pure.
* Every training observation is memorized.

As a result:

* Training accuracy becomes extremely high.
* Test accuracy drops significantly.

This phenomenon is called **Overfitting**.

Therefore, we need to control tree growth using **Hyperparameters**.

---

# 2. Overfitting vs Underfitting

## A. Overfitting (High Variance)

### Definition

A model that performs:

* Very well on Training Data
* Poorly on Test Data

because it memorizes noise and outliers.

### Tree Intuition

Suppose:

```text id="6y9cye"
140 rows
     ↓
split
     ↓
2 rows
```

If a prediction is made based on only 2 samples, then:

* Noise has a huge influence.
* The model becomes unstable.
* Generalization becomes poor.

### Geometric Intuition

The decision boundary creates:

* Tiny boxes
* Small isolated regions
* Highly complex boundaries

to capture every outlier.

---

## Signs of Overfitting

* Low Bias
* High Variance
* Complex Decision Boundary
* High Training Accuracy
* Low Testing Accuracy

---

# B. Underfitting (High Bias)

### Definition

The model is too simple to learn the underlying pattern of the data.

As a result:

* Training Accuracy is low.
* Testing Accuracy is low.

### Tree Intuition

Suppose:

```text id="6bjlwm"
max_depth = 1
```

The tree stops too early.

The resulting leaf nodes still contain heavily mixed classes.

The model cannot capture non-linear patterns.

### Geometric Intuition

The decision boundary becomes:

* Very simple
* Very smooth
* Unable to capture data complexity

---

## Signs of Underfitting

* High Bias
* Low Variance
* Low Training Accuracy
* Low Testing Accuracy

---

# 3. Hyperparameters of Decision Trees

---

# I. criterion

Determines how impurity is measured.

### Options

```python id="nq4gu4"
criterion='gini'
criterion='entropy'
```

### Gini

* Faster computation
* Default in Scikit-Learn

### Entropy

* Uses logarithms
* Sometimes produces cleaner splits

### Practical Observation

Accuracy difference is usually very small.

---

# II. splitter

Determines how candidate splits are selected.

### Options

```python id="vr2wvo"
splitter='best'
splitter='random'
```

---

### splitter='best'

Checks every possible split.

Produces the best split.

May overfit.

---

### splitter='random'

Randomly selects candidate splits.

Advantages:

* Faster training
* Reduces overfitting
* Better generalization

---

# III. max_depth

Controls the maximum depth of the tree.

Default:

```python id="0w8jws"
max_depth=None
```

which means:

```text id="ex95qf"
Grow until all leaves become pure.
```

---

## Small Value

Example:

```python id="ztv0o5"
max_depth=1
```

Results:

* High Bias
* Underfitting

---

## Large Value

Results:

* High Variance
* Overfitting

---

# Main Intuition

The deeper the tree becomes:

* More complex patterns are learned.
* More noise is memorized.

---

# IV. min_samples_split

Minimum number of samples required before a node can split.

Example:

```python id="44ch93"
min_samples_split=100
```

A node containing:

```text id="k09ejl"
73 rows
```

cannot split because:

```text id="h6shha"
73 < 100
```

---

## Larger Value

* Stops tree growth early
* Reduces overfitting

---

## Smaller Value

* Allows deeper trees
* Increases overfitting

---

# V. min_samples_leaf

Minimum number of samples allowed inside a leaf node.

Example:

```python id="gcln1t"
min_samples_leaf=20
```

Every leaf must contain at least:

```text id="j8l1o0"
20 samples
```

Otherwise the split is discarded.

---

## Larger Value

* Creates larger leaf nodes
* Prevents tiny regions
* Reduces overfitting

---

## Smaller Value

* Creates small leaf nodes
* Can memorize noise

---

# VI. max_features

Limits the number of features considered during splitting.

Suppose:

```text id="xikc0n"
100 columns
```

and

```python id="rslcda"
max_features=50
```

At each node:

* Randomly choose 50 columns.
* Find the best split only within those columns.

---

## Why is it useful?

Introducing randomness:

* Reduces overfitting
* Improves generalization

This idea is heavily used in:

```text id="hlcejlwm"
Random Forest
```

---

# VII. max_leaf_nodes

Limits the total number of leaf nodes.

Example:

```python id="yofrc2"
max_leaf_nodes=5
```

The tree can contain at most:

```text id="md6lf0"
5 leaf nodes
```

regardless of depth.

---

## Smaller Value

* Simpler tree
* Less overfitting

---

## Larger Value

* More complex tree
* Higher variance

---

# VIII. min_impurity_decrease

A split occurs only if impurity decreases by at least this amount.

Example:

```python id="yz9txe"
min_impurity_decrease=0.1
```

Small impurity improvements are ignored.

---

## Larger Value

* Early stopping
* Simpler tree
* Reduces overfitting

---

## Smaller Value

* Allows more splits
* Complex tree
* Increases overfitting

---

# Geometric Intuition of Hyperparameters

## Overfitted Tree

Decision boundary:

```text id="fpmq9d"
Many tiny boxes
Many isolated regions
Complex boundaries
```

---

## Underfitted Tree

Decision boundary:

```text id="qmn9ye"
Very few large boxes
Simple boundaries
Misses important patterns
```

---

# Practical Rule of Thumb

When the tree is:

### Overfitting

Make the tree smaller.

When the tree is:

### Underfitting

Allow the tree to grow.

---

# Summary Cheat Sheet

| Hyperparameter          | Reduce Overfitting | Reduce Underfitting |
| ----------------------- | ------------------ | ------------------- |
| `max_depth`             | Decrease           | Increase            |
| `min_samples_split`     | Increase           | Decrease            |
| `min_samples_leaf`      | Increase           | Decrease            |
| `max_features`          | Decrease           | Increase            |
| `max_leaf_nodes`        | Decrease           | Increase            |
| `min_impurity_decrease` | Increase           | Decrease            |
| `splitter`              | Use `'random'`     | Use `'best'`        |

---

| Hyperparameter          | Purpose                                   |
| ----------------------- | ----------------------------------------- |
| `criterion`             | Measure impurity                          |
| `splitter`              | How to search for splits                  |
| `max_depth`             | Maximum tree depth                        |
| `min_samples_split`     | Minimum samples required before splitting |
| `min_samples_leaf`      | Minimum samples allowed in a leaf         |

# Final Takeaway

Decision Trees naturally have a very high tendency to overfit.

Hyperparameters act as **constraints** that control tree growth and improve generalization.

Understanding these parameters is extremely important because they form the foundation of advanced ensemble methods such as:

* Random Forest
* Extra Trees
* Gradient Boosted Decision Trees (GBDT)
* XGBoost
* LightGBM
* CatBoost
