# Visualizing Decision Trees with `dtreeviz`

## Why Do We Need `dtreeviz`?

Scikit-Learn provides a basic way to visualize Decision Trees using `plot_tree()`, but the visualizations often become:

- Difficult to interpret for large trees.
- Less intuitive for understanding split logic.
- Hard to track the path taken by a new observation.
- Limited in showing how data is distributed at each node.

To solve these problems, we can use **dtreeviz**, a powerful visualization library specifically designed for Decision Trees.

---

# What is `dtreeviz`?

`dtreeviz` is an open-source Python library that creates highly interactive and intuitive visualizations for:

- Decision Tree Classifiers
- Decision Tree Regressors

Its main goal is to improve the **interpretability** and **transparency** of tree-based models.

---

# Key Features of `dtreeviz`

## 1. Clear Decision Boundaries

Unlike standard tree plots, `dtreeviz` shows:

- The exact splitting threshold.
- Histograms or scatter plots of the training data.
- How observations are distributed before and after a split.

This makes it much easier to understand **why a split was chosen**.

---

## 2. Supports Regression Trees

`dtreeviz` works for both:

- Classification problems
- Regression problems

For regression trees, it can visualize:

- The target values in each region.
- The mean prediction inside every leaf node.
- How each split changes the prediction.

---

## 3. Path Tracking

One of the most useful features is the ability to visualize:

> The exact path that a new observation follows through the tree.

This helps answer questions such as:

- Why did the model make this prediction?
- Which conditions were satisfied?
- Which features influenced the prediction?

This is extremely useful for:

- Debugging models
- Explaining predictions
- Building intuition

---

## 4. Customization Options

`dtreeviz` provides several customization options:

- Fancy visualizations
- Node labels
- Displaying a specific prediction path
- Different color schemes
- Showing or hiding additional information

---

## 5. Feature Importance Visualization

The library also provides built-in tools to visualize:

- Feature importance rankings
- Contribution of each feature to the model

This makes model interpretation significantly easier.

---

# Advanced Visualizations

`dtreeviz` can also:

- Plot the complete decision tree.
- Visualize decision boundaries.
- Display how the feature space is partitioned.

These visualizations are especially useful for:

- Small datasets
- Educational purposes
- Understanding complex splitting patterns

---

# Why is `dtreeviz` Important?

Decision Trees are highly interpretable models, but their default visualizations often fail to communicate their true logic.

`dtreeviz` helps us:

- Build intuition.
- Understand splits.
- Explain predictions.
- Debug models.
- Improve model transparency.

---

# Main Intuition

> A Decision Tree is simply a collection of if-else rules.

`dtreeviz` allows us to **see those rules visually**, making the entire decision-making process much easier to understand.

````python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from dtreeviz import model 
from IPython.display import Image, display, SVG
import graphviz

import graphviz.backend as be 
````

````python
# Older version 

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Load the Iris dataset
iris = load_iris()

# Features and target
X_train = iris.data
y_train = iris.target

# Create and train the model
clas = DecisionTreeClassifier(random_state=42)
clas.fit(X_train, y_train)

# Plot the decision tree
plt.figure(figsize=(15, 10))
plot_tree(
    clas,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    rounded=True,
    fontsize=10
)

plt.show()
````
