# Introduction to Ensemble Learning

## 1. What is Ensemble Learning?

- The word **Ensemble** means a collection or group of components working together.
- In Machine Learning, Ensemble Learning combines multiple machine learning models (called **base learners** or **weak learners**) to create a stronger and more robust predictive model.
- Ensemble methods are among the most widely used techniques in industry and competitive machine learning because they often outperform individual models.

### Core Idea

Instead of relying on a single model:

- Train multiple models.
- Combine their predictions.
- Produce a final prediction that is generally more accurate and stable.

---

## 2. Wisdom of the Crowd

The foundation of ensemble learning comes from the concept known as the **Wisdom of the Crowd**.

### Principle

A diverse group of individuals often makes better decisions collectively than a single expert.

### Real-World Examples

#### Audience Poll (KBC)

In quiz shows, the audience poll often gives the correct answer because many independent opinions are aggregated together.

#### Product Reviews

A product with:

- Rating: 4.5/5
- Reviews: 15,000+

is generally more trustworthy than:

- Rating: 4.5/5
- Reviews: 1

because the rating represents the opinion of a large crowd.

#### Democracy

Democratic systems rely on collective decisions rather than individual opinions.

#### Ox Weight Experiment

In a famous experiment, hundreds of people guessed the weight of an ox.

- Individual guesses varied significantly.
- The average of all guesses was extremely close to the true weight.

This demonstrates the power of collective intelligence.

---

## 3. Why Diversity is Important

For ensemble learning to work effectively, the individual models must be **diverse**.

### Why?

If all models make the same mistakes, combining them provides no benefit.

The goal is:

- Different models
- Different viewpoints
- Different mistakes

When predictions are aggregated, many errors cancel out.

### Example

Suppose an audience consists only of software engineers.

They may perform well on:

- Programming
- Technology

But poorly on:

- History
- Medicine
- Literature

A diverse audience containing people from multiple backgrounds performs better overall.

---

## 4. How to Create Diversity in Ensemble Models

### Method 1: Different Algorithms

Train different algorithms on the same dataset.

Examples:

- Logistic Regression
- Decision Tree
- Support Vector Machine
- KNN

Each algorithm learns differently and therefore makes different errors.

---

### Method 2: Different Data Subsets

Use the same algorithm but train each model on different subsets of data.

Example:

- Decision Tree 1 → Dataset A
- Decision Tree 2 → Dataset B
- Decision Tree 3 → Dataset C

This creates diversity despite using the same algorithm.

---

### Method 3: Hybrid Approach

Combine:

- Different algorithms
- Different data subsets

This often creates the highest level of diversity.

---

## 5. Prediction Mechanism in Ensemble Learning

After all base models are trained, their predictions are combined.

### Classification Problems

For classification, ensembles typically use **Majority Voting**.

#### Example

Suppose 5 models predict:

| Model | Prediction |
|---------|------------|
| Model 1 | Yes |
| Model 2 | Yes |
| Model 3 | No |
| Model 4 | Yes |
| Model 5 | No |

### Vote Count

| Class | Votes |
|---------|---------|
| Yes | 3 |
| No | 2 |

Final Prediction:

**Yes**

---

### Regression Problems

For regression, ensembles typically use **Mean Averaging**.

#### Example

Predictions from different models:

| Model | Prediction |
|---------|------------|
| Model 1 | 8.1 |
| Model 2 | 3.2 |
| Model 3 | 4.1 |
| Model 4 | 7.5 |

Final Prediction:

$$
\frac{8.1 + 3.2 + 4.1 + 7.5}{4}
=
5.725
$$
Final Output = **5.725**

## 6. Types of Ensemble Techniques

### Overview

```
Ensemble Learning
│
├── Voting
├── Stacking
├── Bagging
└── Boosting
```

---

# 7. Voting Ensemble

## Concept

Voting combines different machine learning algorithms trained on the same dataset.

### Workflow

```
  Same Dataset
       │
 ┌─────┼─────┐
 │     │     │
SVM    LR   DT
 │     │     │
 └─────┼─────┘
       │
   Majority Vote
       │
 Final Prediction
```

### Characteristics

- Different algorithms
- Same dataset
- Independent training
- Parallel execution

### Classification

Uses:

- Majority Voting

### Regression

Uses:

- Mean Averaging

---

# 8. Stacking Ensemble

## Concept

Stacking introduces a second-level model called a **Meta Model**.

Instead of directly combining predictions, a new model learns how to combine them.

### Workflow

```
             Same Dataset
                   │
      ┌────────────┼────────────┐
      │            │            │
     SVM          LR           DT
      │            │            │
      └──────┬─────┴─────┬──────┘
             │           │
            Base Predictions
                  │
            Meta Dataset
                  │
             Meta Model
                  │
           Final Prediction
```

### Benefits

- Learns which models are more reliable.
- Automatically assigns importance to each model.
- Often performs better than simple voting.

---

# 9. Bagging (Bootstrap Aggregation)

## Concept

Bagging uses:

- Same algorithm
- Different bootstrapped datasets

### Bootstrapping

Randomly sample rows **with replacement**.

Example:

Original dataset:

| Rows |
|--------|
| 1000 |

Bootstrapped datasets:

| Model | Training Rows |
|---------|--------------|
| Model 1 | Random Sample D1 |
| Model 2 | Random Sample D2 |
| Model 3 | Random Sample D3 |

Each dataset is slightly different.

### Workflow

```
Original Dataset
        │
 ┌──────┼──────┐
 │      │      │
D1     D2     D3
 │      │      │
DT     DT     DT
 │      │      │
 └──────┼──────┘
        │
 Aggregate
        │
 Final Prediction
```

### Advantages

- Reduces variance
- Reduces overfitting
- More stable predictions

### Special Case

**Random Forest = Bagging + Decision Trees**

Random Forest is the most famous bagging algorithm.

---

# 10. Boosting

## Concept

Boosting trains models sequentially.

Each new model attempts to correct the mistakes made by previous models.

### Workflow

```
Dataset
   │
Model 1
   │
Errors
   │
Model 2
   │
Remaining Errors
   │
Model 3
   │
Final Prediction
```

### Key Idea

Every subsequent model focuses more on difficult observations.

### Popular Boosting Algorithms

1. AdaBoost
2. Gradient Boosting
3. XGBoost
4. LightGBM
5. CatBoost

---

## 11. Noise Cancellation Effect

Ensemble methods improve performance through error cancellation.

### Classification

Individual models may create:

- Noisy boundaries
- Overfitted regions
- Local mistakes

Combining multiple boundaries creates a smoother and more generalized final decision boundary.

### Regression

Outlier predictions are balanced by averaging.

Example:

| Model | Prediction |
|---------|------------|
| Model 1 | 5 |
| Model 2 | 6 |
| Model 3 | 5.5 |
| Model 4 | 20 |

Average prediction reduces the influence of extreme values.

---

## 12. Bias-Variance Tradeoff

### Traditional Problem

Reducing bias often increases variance.

Reducing variance often increases bias.

### How Ensembles Help

Ensemble methods help break this tradeoff.

---

### Bagging

Best for:

- High Variance Models
- Low Bias Models

Example:

- Deep Decision Trees

Bagging reduces variance while preserving low bias.

---

### Boosting

Best for:

- High Bias Models
- Low Variance Models

Example:

- Decision Stumps

Boosting reduces bias while keeping variance under control.

---

## 13. Advantages of Ensemble Learning

### 1. Higher Accuracy

Usually outperforms individual models.

### 2. Better Generalization

Works better on unseen data.

### 3. Reduced Overfitting

Especially with Bagging and Random Forest.

### 4. More Robust

Less sensitive to noise.

### 5. Handles Complex Patterns

Can learn highly non-linear relationships.

---

## 14. Disadvantages of Ensemble Learning

### 1. Increased Computational Cost

Multiple models require:

- More memory
- More CPU time
- More storage

### 2. Slower Training

Training many models takes longer.

### 3. Lower Interpretability

Harder to explain compared to a single Decision Tree.

### 4. More Complex Deployment

Managing multiple models is more challenging.

---

## 15. When Should You Use Ensemble Learning?

Recommended workflow:

1. Data Cleaning
2. Exploratory Data Analysis
3. Feature Engineering
4. Train Baseline Models
5. Hyperparameter Tuning
6. Apply Ensemble Techniques

Ensemble methods should generally be used after establishing strong baseline models.

## Key Takeaways

- Ensemble Learning combines multiple models to create a stronger predictor.
- Diversity among models is essential.
- Classification uses voting, while regression uses averaging.
- Four major techniques are:
  - Voting
  - Stacking
  - Bagging
  - Boosting
- Random Forest is a Bagging algorithm.
- XGBoost is a Boosting algorithm.
- Bagging reduces variance.
- Boosting reduces bias.
- Ensemble methods are among the most powerful techniques for tabular machine learning datasets.
