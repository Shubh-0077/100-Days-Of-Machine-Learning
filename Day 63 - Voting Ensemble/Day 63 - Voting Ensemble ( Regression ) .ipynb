```python
import numpy as np 
from sklearn.datasets import fetch_california_housing
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor 
from sklearn.svm import SVR  
from sklearn.model_selection import cross_val_score  
```

```python
x,y = fetch_california_housing(return_X_y=True)
```

```python
x.shape
```

```python
y.shape
```

```python
x
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

```python
estimators = [('lr',lr),('dt',dt),('svr',svr)] 
```

```python
# Lets train our multiple Model 
for estimator in estimators:
    score = cross_val_score(estimator[1],x,y,scoring='r2',cv=10)

    print(estimator[0], np.round(score.mean(), 3))
```

```python
# Lets Do Voting , and see how our model's performance improves 
from sklearn.ensemble import VotingRegressor 
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,x,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```

```python
#  Lets add weight to the model , but first lets see which weights combination works best 
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,x,y,scoring='r2',cv=10)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
```

```python

# using the same algorithm but changing hyperparameters

dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],x,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,x,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```
