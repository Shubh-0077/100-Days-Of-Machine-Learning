# Feature Importance

## 1. What is Feature Importance?

Feature Importance assigns a numerical score to every input feature in a dataset and indicates how much the model relies on that feature while making predictions.

### Why is Feature Importance Important?

| Use Case                 | Description                                                                                                 |
| ------------------------ | ----------------------------------------------------------------------------------------------------------- |
| Feature Selection        | Helps identify the most useful features and remove irrelevant ones, reducing training time and overfitting. |
| Model Interpretability   | Explains which features influenced model predictions the most.                                              |
| Dimensionality Reduction | Useful when working with datasets containing hundreds or thousands of features.                             |
| Business Insights        | Helps understand the key factors driving model decisions.                                                   |

---

## 2. Intuition Using the MNIST Dataset

Consider the MNIST handwritten digit dataset:

* Each image consists of **28 × 28 = 784 pixels**.
* Every pixel acts as a separate feature.
* Corner pixels are usually blank and contribute very little information.
* Central pixels contain the actual digit strokes and are highly informative.

When Feature Importance is visualized as a heatmap:

* Bright regions indicate highly important pixels.
* Dark regions indicate unimportant pixels.

This helps identify which parts of an image contribute most to classification.

# 3. Feature Importance in a Decision Tree

Decision Trees calculate Feature Importance using:

* **Gini Importance**
* **Mean Decrease in Impurity (MDI)**

The core idea is:

> A feature is considered important if it reduces impurity significantly across multiple splits.

---

## Step 1: Calculate Node Importance

For every internal node (k), we calculate the Node Importance:

$$
NI_k
=
\frac{N_k}{N}
\times
Impurity_k
-
\left(
\frac{N_{kL}}{N}
\times
Impurity_{kL}
+
\frac{N_{kR}}{N}
\times
Impurity_{kR}
\right)
$$
-
### Where:

| Symbol          | Meaning                     |
| --------------- | --------------------------- |
| (N)             | Total training samples      |
| (N_k)           | Samples reaching node (k)   |
| (N_{kL})        | Samples in left child node  |
| (N_{kR})        | Samples in right child node |
| (Impurity_k)    | Impurity at node (k)        |
| (Impurity_{kL}) | Impurity of left child      |
| (Impurity_{kR}) | Impurity of right child     |

### Interpretation

The formula measures:

> How much impurity decreases after performing the split.

A larger decrease means a more useful split.

---

## Step 2: Aggregate Importance for Each Feature

For a feature (i):

* Find all nodes where feature (i) was used for splitting.
* Sum the Node Importance values of those nodes.

$$
FI_i
=
\sum_{k \in \text{Nodes using Feature } i}
NI_k
$$
---

## Step 3: Normalize Feature Importance

To make the scores easier to interpret, Scikit-Learn normalizes them so their sum equals 1.

$$
FeatureImportance_i
=
\frac{
\sum_{k \in \text{Nodes using Feature } i}
NI_k
}
{
\sum_{j \in \text{All Internal Nodes}}
NI_j
}
$$
-
### Key Property

$$
\sum_{i=1}^{p}
FeatureImportance_i
=
1
$$
-
where (p) is the total number of features.

---

# 4. Feature Importance in Random Forest

A Random Forest consists of multiple Decision Trees.

The process is simple:

### Step 1

Calculate Feature Importance for every tree individually.

### Step 2

Average the importance scores across all trees.

If there are (M) trees:

$$
RFImportance_i
=
\frac{1}{M}
\sum_{t=1}^{M}
FeatureImportance_{i,t}
$$
-
where:

* (M) = Number of trees
* (FeatureImportance_{i,t}) = Importance of feature (i) in tree (t)

---

## Example

Suppose a Random Forest contains only two trees.

### Tree 1

| Feature | Importance |
| ------- | ---------- |
| X₀      | 1.0        |
| X₁      | 0.0        |

### Tree 2

| Feature | Importance |
| ------- | ---------- |
| X₀      | 0.4        |
| X₁      | 0.6        |

### Final Random Forest Importance

For (X_0):

$$
Importance(X_0)
=
# \frac{1.0 + 0.4}{2}
0.70
$$
-

For (X_1):

$$
Importance(X_1)
==

# \frac{0.0 + 0.6}{2}

0.30
$$
-
### Final Result

| Feature | Importance |
| ------- | ---------- |
| X₀      | 0.70       |
| X₁      | 0.30       |

This means the Random Forest relies more heavily on (X_0).

# 5. Important Limitation: High Cardinality Bias

Feature Importance based on impurity reduction is not perfect.

### Problem

Tree-based models naturally prefer:

* Numerical columns
* Features with many unique values
* IDs
* Timestamps

These features provide many possible split points.

As a result:

> The model may assign them artificially high importance even when they have little predictive power.

---

## Example

A customer ID column may appear highly important simply because every value is unique.

However:

* It does not truly explain customer behavior.
* The importance score becomes misleading.

---

# 6. Solution: Permutation Importance

When working with high-cardinality features, use:

**Permutation Importance**

instead of impurity-based importance.

### Why?

Permutation Importance measures:

> How much model performance drops when a feature's values are randomly shuffled.

If performance drops significantly:

* The feature is important.

If performance barely changes:

* The feature is not important.

This produces a more reliable and unbiased estimate of feature importance.

---

# Key Takeaways

* Feature Importance measures how much a model relies on each feature.
* Decision Trees use **Mean Decrease in Impurity (MDI)**.
* Node Importance is calculated from impurity reduction after a split.
* Feature Importance is the normalized sum of all node importances where that feature was used.
* Random Forest Feature Importance is the average importance across all trees.
* Impurity-based importance can be biased toward high-cardinality features.
* **Permutation Importance** is often preferred when interpretability is critical.
