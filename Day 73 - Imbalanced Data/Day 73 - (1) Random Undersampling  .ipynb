# Random Undersampling

**Random Undersampling** is one of the simplest techniques used to handle **imbalanced datasets**.

Instead of increasing the number of minority class samples, Random Undersampling balances the dataset by **reducing the number of majority class samples**.

The idea is straightforward:

> If one class has significantly more observations than the other, randomly remove samples from the majority class until both classes have a similar number of observations.

This creates a balanced dataset that allows the machine learning model to give equal importance to both classes during training.

---

# 1. Why Do We Need Random Undersampling?

Consider the following dataset.

```
Total Customers = 10,000

Purchased Product = 1,000

Did Not Purchase = 9,000
```

Here,

```
Majority Class = Did Not Purchase

Minority Class = Purchased Product
```

A machine learning model trained on this data will mostly see "Did Not Purchase" examples.

As a result,

it becomes biased toward predicting the majority class.

Random Undersampling solves this problem by reducing the majority class size.

---

# 2. Core Intuition

Imagine a classroom.

```
90 Boys

10 Girls
```

Suppose you want to study the behavior of both groups equally.

Instead of collecting more girls,

you randomly select only **10 boys**.

Now you have

```
10 Boys

10 Girls
```

Both groups now receive equal attention.

Random Undersampling follows the same idea.

---

# 3. How Random Undersampling Works

Suppose we have the following dataset.

```
Majority Class

■■■■■■■■■■■■■■■■■■■■

20 Samples

Minority Class

■■■■■

5 Samples
```

The algorithm randomly removes majority class samples.

After undersampling,

```
Majority Class

■■■■■

5 Samples

Minority Class

■■■■■

5 Samples
```

Now,

both classes contain the same number of observations.

---

# 4. Step-by-Step Workflow

## Step 1

Count the number of observations in each class.

Example

```
Class A = 900

Class B = 100
```

---

## Step 2

Identify the majority class.

```
Majority = Class A

Minority = Class B
```

---

## Step 3

Randomly select observations from the majority class.

Only keep enough samples so that both classes become balanced.

Example

```
Keep only

100 samples from Class A
```

---

## Step 4

Combine the selected majority samples with all minority samples.

Final dataset

```
Class A = 100

Class B = 100
```

---

## Step 5

Train the machine learning model using this balanced dataset.

---

# 7. Advantages

### Simple to Implement

One of the easiest balancing techniques.

---

### Faster Training

Since many samples are removed,

the dataset becomes much smaller.

Smaller datasets require

- less memory
- less computation
- faster training

---

### Reduces Majority Bias

The classifier receives equal representation from both classes.

This helps reduce prediction bias toward the majority class.

---

### Useful for Very Large Datasets

When millions of majority samples exist,

removing some of them usually does not significantly affect performance.

---

# 8. Limitations

### Information Loss

The biggest drawback is that useful information may be discarded.

Since samples are removed randomly,

important patterns may disappear.

Example

Suppose one removed transaction represents a unique fraud pattern.

After deletion,

the model will never learn that pattern.

---

### Lower Model Performance

Removing too much data may reduce the model's ability to generalize.

---

### Random Selection

The algorithm has no knowledge of which samples are important.

It removes observations purely at random.

---

### Not Suitable for Small Datasets

If the dataset is already small,

removing observations can significantly reduce model performance.

---

# 9. When Should You Use Random Undersampling?

Random Undersampling works well when

- The dataset is extremely large.
- The majority class contains many redundant observations.
- Training speed is important.
- Slight information loss is acceptable.

Avoid it when

- The dataset is small.
- Every observation is valuable.
- Losing information could affect prediction quality.

---

# 10. Best Practices

- Apply undersampling **only on the training dataset**.
- Never undersample the test dataset.
- Always evaluate using Precision, Recall, and F1 Score instead of Accuracy.
- Compare Random Undersampling with other balancing techniques before selecting the final model.

To reduce the number of majority class samples and create a balanced dataset.

No.

# Key Takeaway

Random Undersampling is the **simplest balancing technique** for imbalanced datasets. It works by **randomly removing majority class samples**, allowing the model to learn from a balanced dataset.

While it is computationally efficient and reduces majority class bias, its biggest disadvantage is **information loss**, making it more suitable for **large datasets** where removing some observations does not significantly affect the overall data distribution.

````python
import numpy as np 
import pandas as pd  
import matplotlib.pyplot as plt 

from sklearn.datasets import make_blobs 

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report , roc_auc_score, roc_curve

from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

import xgboost as xgb  
````

````python
# Set random seed for reproducibility
np.random.seed(42)

# Generate imbalanced dataset
n_samples_1 = 25  # Number of samples in class 1
n_samples_2 = 375  # Number of samples in class 2
centers = [(0, 0), (2, 2)]  # Centers of each cluster
cluster_std = [1.5, 1.5]  # Standard deviation of each cluster

x, y = make_blobs(n_samples=[n_samples_1, n_samples_2],
                  centers=centers,
                  cluster_std=cluster_std,
                  random_state=0)
````

````python
from imblearn.under_sampling import RandomUnderSampler

# Split dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Applying Random Over Sampling
rus = RandomUnderSampler(random_state=42)
x_resampled, y_resampled = rus.fit_resample(x_train, y_train)

# Plotting
plt.figure(figsize=(8, 6))
plt.scatter(x_train[y_train == 0][:, 0], x_train[y_train == 0][:, 1], color='red', label='Class 1')
plt.scatter(x_train[y_train == 1][:, 0], x_train[y_train == 1][:, 1], color='blue', label='Class 2')
plt.title('2D Imbalanced Dataset with Two Classes')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()

# Plotting
plt.figure(figsize=(8, 6))
plt.scatter(x_resampled[y_resampled == 0][:, 0], x_resampled[y_resampled == 0][:, 1], color='red', label='Class 1')
plt.scatter(x_resampled[y_resampled == 1][:, 0], x_resampled[y_resampled == 1][:, 1], color='blue', label='Class 2')
plt.title('2D Imbalanced Dataset with Two Classes')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
````
