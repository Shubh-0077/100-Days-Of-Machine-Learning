# Random Oversampling

**Random Oversampling** is one of the most widely used techniques for handling **imbalanced datasets**.

Unlike **Random Undersampling**, which removes majority class samples, Random Oversampling balances the dataset by **increasing the number of minority class samples**.

It does this by **randomly duplicating existing minority class observations** until both classes contain approximately the same number of samples.

The goal is simple:

> Instead of deleting useful majority class data, duplicate minority class samples so that the machine learning model learns equally from both classes.

---

# 1. Why Do We Need Random Oversampling?

Consider the following dataset.

```
Credit Card Transactions

Legitimate = 9,000

Fraud = 1,000
```

Here,

```
Majority Class = Legitimate

Minority Class = Fraud
```

A machine learning model trained on this dataset mainly sees legitimate transactions.

As a result,

it learns the characteristics of legitimate transactions much better than fraudulent ones.

Random Oversampling solves this problem by increasing the number of minority class samples.

---

# 2. Core Intuition

Imagine a classroom.

```
90 Boys

10 Girls
```

Instead of removing boys,

you simply ask each girl to appear multiple times in the practice sessions.

Eventually,

the classroom becomes

```
90 Boys

90 Girls
```

Now,

both groups receive equal attention during learning.

Random Oversampling works exactly like this.

It does **not create new data**.

It simply **duplicates existing minority samples**.

---

# 3. How Random Oversampling Works

Suppose our dataset contains

```
Majority Class

■■■■■■■■■■■■■■■■■■■■

20 Samples

Minority Class

■■■■■

5 Samples
```

The algorithm randomly selects minority samples and duplicates them.

Example

```
Original Minority

A

B

C

D

E
```

After Oversampling

```
A

B

C

D

E

B

C

A

E

D

...
```

Eventually,

```
Majority = 20

Minority = 20
```

Both classes become balanced.

---

# 4. Step-by-Step Workflow

## Step 1

Count the number of samples in each class.

Example

```
Class A = 900

Class B = 100
```

---

## Step 2

Identify the minority class.

```
Minority = Class B
```

---

## Step 3

Randomly choose minority class samples.

These samples are selected with replacement.

This means

the same observation can be selected multiple times.

---

## Step 4

Duplicate the selected observations until the minority class reaches the desired size.

Example

```
Class A = 900

Class B = 900
```

---

## Step 5

Train the machine learning model using the balanced dataset.

# 6. Example

Suppose we have

```
Loan Dataset

Loan Approved = 9,000

Loan Default = 1,000
```

Random Oversampling repeatedly duplicates loan default cases.

Result

```
Loan Approved = 9,000

Loan Default = 9,000
```

The classifier now receives equal examples from both classes.

---

# 7. Advantages

### No Information Loss

Unlike Random Undersampling,

no observations are removed.

Every original sample remains in the dataset.

---

### Better Minority Learning

Since minority samples appear more frequently,

the model learns their characteristics much better.

---

### Simple to Implement

Random Oversampling is easy to understand and implement.

---

### Useful for Small Datasets

When the dataset is small,

removing observations can be harmful.

Oversampling preserves all available information.

---

# 8. Limitations

### Overfitting

This is the biggest drawback.

Since the algorithm simply duplicates existing samples,

the model repeatedly sees identical observations.

Instead of learning generalized patterns,

it may simply memorize those duplicated samples.

---

### Larger Dataset

Oversampling increases the total number of observations.

As a result,

- training becomes slower,
- memory usage increases,
- computational cost increases.

---

### No New Information

Duplicated samples do not introduce any new knowledge.

They only increase the frequency of existing observations.

---

# 9. Random Undersampling vs Random Oversampling

| Random Undersampling | Random Oversampling |
|----------------------|--------------------|
| Removes majority samples | Duplicates minority samples |
| Dataset becomes smaller | Dataset becomes larger |
| Faster training | Slower training |
| Information loss | No information loss |
| Lower memory usage | Higher memory usage |
| Risk of losing patterns | Risk of overfitting |

---

# 10. When Should You Use Random Oversampling?

Random Oversampling works well when

- The dataset is relatively small.
- Losing data is not acceptable.
- Minority class detection is important.
- Training time is not a major concern.

Avoid it when

- The dataset is already extremely large.
- Duplicate observations may lead to severe overfitting.

---

# 11. Best Practices

- Apply oversampling **only on the training dataset**.
- Never oversample the testing dataset.
- Compare Random Oversampling with SMOTE.
- Use Precision, Recall, and F1 Score for evaluation instead of Accuracy.

# Key Takeaway

Random Oversampling is one of the simplest techniques for handling imbalanced datasets. It balances the data by **duplicating existing minority class samples**, allowing the model to learn equally from both classes.

Although it preserves all original information, its major limitation is **overfitting**, since the duplicated samples do not provide any new information to the model. This limitation motivated more advanced techniques like **SMOTE**, which generates **synthetic minority samples instead of exact duplicates**.
