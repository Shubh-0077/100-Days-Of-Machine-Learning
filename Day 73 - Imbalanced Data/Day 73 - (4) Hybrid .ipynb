# Hybrid Methods for Handling Imbalanced Data

Hybrid Methods combine **Oversampling** and **Undersampling** techniques to take advantage of both approaches while reducing their individual drawbacks.

Instead of only generating new minority samples or only removing majority samples, Hybrid Methods first create synthetic minority data and then clean the dataset by removing noisy or overlapping samples.

This often produces cleaner decision boundaries and better model performance.

---

# 1. Why Do We Need Hybrid Methods?

Let's recall the problems with the previous techniques.

### Random Oversampling

- Duplicates existing minority samples.
- Can cause overfitting.

### SMOTE

- Generates synthetic samples.
- But it doesn't know where the majority class exists.
- Sometimes synthetic samples are created inside majority regions.

### Undersampling

- Removes majority samples.
- May remove useful information.

Hybrid methods solve these problems by combining the strengths of Oversampling and Undersampling.

---

# 2. How Hybrid Methods Work

The overall workflow is simple.

```
Original Dataset

↓

Oversample Minority Class

↓

Clean Noisy Samples

↓

Balanced Dataset
```

The most popular Hybrid Method is

```
SMOTE + Tomek Links
```

---

# 3. SMOTE + Tomek Links

This method has two phases.

---

## Phase 1 : Apply SMOTE

First,

SMOTE generates synthetic minority samples.

Example

Before SMOTE

```
Majority : 900

Minority : 100
```

After SMOTE

```
Majority : 900

Minority : 900
```

Now both classes are balanced.

However,

some synthetic points may lie very close to the majority class.

This creates overlapping regions.

---

## Phase 2 : Apply Tomek Links

Now the algorithm searches for Tomek Links.

---

# 4. What is a Tomek Link?

Suppose we have two points.

```
A → Majority Class

B → Minority Class
```

If

- A is the nearest neighbor of B
- B is also the nearest neighbor of A

then they form a

```
Tomek Link
```

These pairs usually exist near the class boundary.

Example

```
Majority ●

          ×

Minority ○
```

The two points are extremely close to each other.

This indicates an overlapping region.

---

# 5. Removing Tomek Links

Once Tomek Links are found,

the majority class sample is removed.

Sometimes both samples may be removed depending on the implementation.

Removing these noisy points creates a cleaner separation between the two classes.

```
Before Cleaning

●   ○

↓

After Cleaning

○
```

The decision boundary becomes much clearer.

---

# 6. Advantages

- Produces cleaner datasets.
- Removes noisy observations.
- Creates better decision boundaries.
- Usually performs better than using SMOTE alone.
- Reduces class overlap.

---

# 7. Limitations

- Slightly more computationally expensive.
- Performance depends on SMOTE parameters.
- May remove useful boundary samples if not tuned properly.

---

# 8. When Should You Use Hybrid Methods?

Use Hybrid Methods when

- Classes overlap heavily.
- SMOTE alone creates noisy samples.
- Better decision boundaries are required.
- You want higher model generalization.

---

# 9. Summary

| Concept | Description |

| Learning Type | Data Resampling Technique |
| Main Idea | Oversample first, then clean the data |

| Step 1 | Generate synthetic minority samples |
