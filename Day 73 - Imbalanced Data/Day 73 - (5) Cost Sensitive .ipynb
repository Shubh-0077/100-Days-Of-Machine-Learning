# Cost-Sensitive Learning for Imbalanced Data

Unlike Oversampling and Undersampling techniques, **Cost-Sensitive Learning** does **not modify the dataset**.

Instead, it changes how the machine learning algorithm learns by assigning **higher penalties to mistakes made on the minority class**.

The model becomes more careful while predicting the minority class because making those mistakes becomes more "expensive."

---

# 1. Why Do We Need Cost-Sensitive Learning?

Imagine a fraud detection problem.

```
Legitimate Transactions : 99%

Fraud Transactions : 1%
```

A normal model may simply predict

```
Legitimate
```

for every transaction and achieve

```
99% Accuracy
```

But it completely fails to detect fraud.

In real-world applications,

missing a fraud transaction is much more serious than wrongly flagging a normal transaction.

Cost-Sensitive Learning teaches the model exactly this idea.

---

# 2. Main Idea

Instead of changing the data,

we change the **cost of making mistakes**.

The model receives

- Small penalty for majority mistakes.
- Large penalty for minority mistakes.

```
Majority Error

↓

Small Penalty

-------------------

Minority Error

↓

Large Penalty
```

Because minority mistakes are expensive,

the algorithm tries harder to correctly classify them.

---

# 3. Cost Matrix

The penalty is represented using a Cost Matrix.

Example

```
                 Predicted

             Positive   Negative

Actual

Positive        0         High Cost

Negative     Low Cost        0
```

Here,

misclassifying a positive (minority) sample has a much larger cost.

---

# 4. How Does It Work?

During training,

the loss function is modified.

Instead of treating every error equally,

different weights are assigned to different classes.

```
Minority Class

Higher Weight

↓

Higher Loss

↓

Model Learns Better
```

Many machine learning algorithms allow class weights.

Examples include

- Logistic Regression
- Decision Tree
- Random Forest
- SVM
- XGBoost
- LightGBM

---

# 5. Balanced Random Forest

One popular cost-sensitive ensemble method is

```
Balanced Random Forest
```

Unlike a normal Random Forest,

each tree is trained using a balanced bootstrap sample.

Workflow

```
Original Dataset

↓

Random Sampling

↓

Balanced Sample

↓

Train Decision Tree

↓

Repeat for Multiple Trees
```

Each tree sees approximately equal numbers of majority and minority samples.

As a result,

the overall forest becomes much better at predicting the minority class.

---

# 6. Advantages

- No synthetic data generation.
- No information loss.
- Works well for highly imbalanced datasets.
- Easy to implement using class weights.
- Better generalization than simple oversampling.

---

# 7. Limitations

- Choosing appropriate class weights can be difficult.
- Very high penalties may increase false positives.
- Performance depends on the underlying algorithm.

---

# 8. When Should You Use Cost-Sensitive Learning?

Use Cost-Sensitive Learning when

- Losing data is unacceptable.
- Creating synthetic data is undesirable.
- The dataset is highly imbalanced.
- Minority class prediction is extremely important.
- Working with fraud detection, medical diagnosis, cybersecurity, or anomaly detection.

---

# 9. Summary

| Concept | Description |
|---------|-------------|
| Learning Type | Algorithm-Level Technique |
| Main Idea | Increase the cost of minority class mistakes |
| Data Modification | No |
| Uses Synthetic Data | No |
| Uses Class Weights | Yes |
| Popular Model | Balanced Random Forest |
| Advantage | Learns minority class without changing the dataset |
| Limitation | Requires proper tuning of class weights |

# Overall Workflow

```text
         Original Dataset
                │
                ▼
     Assign Class Weights
                │
                ▼
     Modify Loss Function
                │
                ▼
      Train ML Algorithm
                │
                ▼
   Higher Penalty for Minority Errors
                │
                ▼
      Improved Minority Prediction
```
